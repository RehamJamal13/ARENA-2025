```python
def generate_joint_extraction(sentence, model, tokenizer):
    """Complete pipeline: prompt → model → parsed results for BOTH entities and relations"""
    
    # Show the sentence being processed
    print(f"\n📄 Processing sentence: '{sentence}'")
    print("=" * 80)
    
    # 1. Create the joint extraction prompt using the function
    prompt = create_intelligent_mistral_joint_extraction_prompt(sentence)
    
    print(f"📝 Generated prompt (first 200 chars): {prompt[:200]}...")
    
    # 2. Tokenize the prompt
    inputs = tokenizer(
        prompt, 
        return_tensors="pt", 
        truncation=True, 
        max_length=3000,
        padding=True
    )
    
    # Move to GPU if available
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    
    print(f"🔧 Input tokenized. Length: {inputs['input_ids'].shape[1]} tokens")
    
    # 3. Generate response from model
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=500,
            temperature=0.1,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
            repetition_penalty=1.1
        )
    
    # 4. Decode the generated response
    generated_text = tokenizer.decode(
        outputs[0][inputs['input_ids'].shape[1]:],
        skip_special_tokens=True
    )
    
    print(f"🤖 Model response:\n{generated_text}")
    print("=" * 80)
    
    return generated_text, prompt
```

```python
def parse_joint_model_response(response, original_sentence):
    """Parse model response with better error handling for various formats."""
    
    result = {
        "entities": [],
        "relations": []
    }
    
    lines = response.strip().split('\n')
    current_section = None
    
    print(f"🔍 Parsing {len(lines)} lines...")
    
    for line_num, line in enumerate(lines):
        line = line.strip()
        
        if not line:
            continue
            
        print(f"  Line {line_num}: '{line}'")
        
        # Detect sections
        if 'ENTITIES:' in line.upper():
            current_section = 'entities'
            print("    📋 Found ENTITIES section")
            continue
        elif 'RELATIONS:' in line.upper():
            current_section = 'relations'
            print("    🔗 Found RELATIONS section")
            continue
        
        # Parse entities
        if current_section == 'entities' and '|' in line:
            line = line.lstrip('- ').strip()
            parts = line.split('|')
            if len(parts) >= 2:
                entity_text = parts[0].strip()
                entity_type = parts[1].strip()
                
                # Handle parentheses in entity type (e.g., "LOC, LULC (urban area)")
                if '(' in entity_type:
                    entity_type = entity_type.split(',')[0].strip()
                
                # Fix common issues
                if '%' in entity_text and entity_type == 'CARDINAL':
                    entity_type = 'PERCENT'
                
                start_pos = original_sentence.lower().find(entity_text.lower())
                end_pos = start_pos + len(entity_text) if start_pos != -1 else -1
                
                entity = {
                    "text": entity_text,
                    "label": entity_type,
                    "start_char": start_pos,
                    "end_char": end_pos
                }
                
                result["entities"].append(entity)
                print(f"    ✅ Entity: {entity_text} ({entity_type})")
        
        # Parse relations - FIXED FOR YOUR FORMAT
        elif current_section == 'relations' and '--' in line:
            import re
            
            # Updated patterns to match your model's output
            patterns = [
                # Pattern 1: "- entity1 --RELATIONSHIP-- entity2 | CONF: LEVEL"
                (r'^-?\s*(.+?)\s*--([A-Z_]+)--\s*(.+?)\s*\|\s*CONF:\s*(\w+)', "CONF Format"),
                
                # Pattern 2: With brackets "- entity1 --[RELATIONSHIP]--> entity2 | confidence: LEVEL"
                (r'^-?\s*(.+?)\s*--\[([A-Z_]+)\]-->\s*(.+?)\s*\|\s*confidence:\s*(\w+)', "Bracket Arrow"),
                
                # Pattern 3: Simple arrow without brackets
                (r'^-?\s*(.+?)\s*--([A-Z_]+)-->\s*(.+?)\s*\|\s*confidence:\s*(\w+)', "Simple Arrow"),
                
                # Pattern 4: Without confidence marker
                (r'^-?\s*(.+?)\s*--([A-Z_]+)--\s*(.+?)$', "Simple No Conf"),
            ]
            
            parsed = False
            for pattern, pattern_name in patterns:
                match = re.match(pattern, line)
                if match:
                    source = match.group(1).strip()
                    relationship = match.group(2).strip().upper()
                    target = match.group(3).strip()
                    
                    # Handle confidence
                    if len(match.groups()) >= 4:
                        confidence = match.group(4).strip().upper()
                        # Normalize confidence values
                        if confidence in ['HIGH', 'MEDIUM', 'LOW']:
                            pass  # Already correct
                        elif confidence.isdigit():
                            # Convert percentage to HIGH/MEDIUM/LOW
                            conf_int = int(confidence)
                            confidence = "HIGH" if conf_int >= 80 else "MEDIUM" if conf_int >= 50 else "LOW"
                        else:
                            confidence = "MEDIUM"  # Default
                    else:
                        confidence = "MEDIUM"  # Default if no confidence specified
                    
                    # Skip invalid relations
                    if '--CHANGE--' in target or '--TRANSFORM--' in target:
                        # This is a malformed relation
                        print(f"    ❌ Skipping malformed relation: {line}")
                        continue
                    
                    # Create relation
                    relation = {
                        "source": source,
                        "relationship": relationship,
                        "target": target,
                        "confidence": confidence
                    }
                    
                    result["relations"].append(relation)
                    print(f"    ✅ {pattern_name}: {source} --[{relationship}]--> {target} ({confidence})")
                    parsed = True
                    break
            
            if not parsed:
                print(f"    ❌ Could not parse relation: '{line}'")
                # Debug: Try to understand why
                if '|' not in line:
                    print(f"        Missing pipe separator")
                elif 'CONF:' not in line and 'confidence:' not in line:
                    print(f"        Missing confidence marker")
                else:
                    # Try basic split to see what we're dealing with
                    parts = line.split('--')
                    if len(parts) >= 3:
                        print(f"        Parts: {parts}")
    
    print(f"📊 Total entities parsed: {len(result['entities'])}")
    print(f"🔗 Total relations parsed: {len(result['relations'])}")
    return result
```

```python
def find_best_entity_match(text, entity_id_map):
    """Find best matching entity using fuzzy matching."""
    
    # Simple fuzzy matching strategies
    text_lower = text.lower().strip()
    
    for entity_text, entity_id in entity_id_map.items():
        entity_lower = entity_text.lower().strip()
        
        # Exact match (case insensitive)
        if text_lower == entity_lower:
            return entity_id
        
        # Substring match
        if text_lower in entity_lower or entity_lower in text_lower:
            return entity_id
        
        # Handle common variations
        if text_lower.replace('_', ' ') == entity_lower or entity_lower.replace(' ', '_') == text_lower:
            return entity_id
    
    return None
```

```python
def create_dynamic_label_studio_config_fixed(processing_results, output_dir):
    """Create Label Studio config with better error handling."""
    
    all_relation_types = set()
    all_entity_types = set()
    
    # Extract all unique types
    for result in processing_results:
        for relation in result.get('relations', []):
            all_relation_types.add(relation['relationship'])
        for entity in result.get('entities', []):
            all_entity_types.add(entity['label'])
    
    # Only proceed if we have data
    if not all_entity_types and not all_relation_types:
        print("⚠️ No entity or relation types found for config generation")
        return None
    
    print(f"📊 Found {len(all_entity_types)} entity types, {len(all_relation_types)} relation types")
    
    colors = [
        "#FF6B6B", "#4ECDC4", "#45B7D1", "#96CEB4", "#FECA57", 
        "#FF9FF3", "#A55EEA", "#54A0FF", "#5F27CD", "#00D2D3",
        "#FF9F43", "#10AC84", "#EE5A24", "#0098C7", "#8395A7"
    ]
    
    # Ensure output directory exists
    os.makedirs(output_dir, exist_ok=True)
    
    LABEL_STUDIO_CONFIG_PATH = os.path.join(output_dir, "joint_label_studio_config.xml")
    
    try:
        # Build entity labels with XML escaping
        entity_labels = ""
        for i, entity_type in enumerate(sorted(all_entity_types)):
            color = colors[i % len(colors)]
            # Escape XML special characters
            safe_entity = xml.sax.saxutils.escape(entity_type)
            entity_labels += f'    <Label value="{safe_entity}" background="{color}"/>\n'
        
        # Build relation labels with XML escaping
        relation_labels = ""
        for i, relation_type in enumerate(sorted(all_relation_types)):
            color = colors[i % len(colors)]
            # Escape XML special characters
            safe_relation = xml.sax.saxutils.escape(relation_type)
            relation_labels += f'    <Relation value="{safe_relation}" background="{color}"/>\n'
        
        # Build complete XML config
        label_studio_config = f"""<View>
  <Text name="text" value="$text"/>
  <Labels name="label" toName="text">
{entity_labels}  </Labels>
  <Relations name="relation" toName="label">
{relation_labels}  </Relations>
</View>"""
        
        # Write the file
        with open(LABEL_STUDIO_CONFIG_PATH, 'w', encoding='utf-8') as f:
            f.write(label_studio_config)
        
        print(f"✅ Label Studio config saved to: {LABEL_STUDIO_CONFIG_PATH}")
        
        return LABEL_STUDIO_CONFIG_PATH
        
    except Exception as e:
        print(f"❌ Error creating Label Studio config: {e}")
        return None
```

```python
def save_joint_extraction_results_and_create_label_studio(processing_results, output_dir="output"):
    """
    Save joint extraction results and create Label Studio files.
    """
    import os
    
    os.makedirs(output_dir, exist_ok=True)
    
    JOINT_OUTPUT = os.path.join(output_dir, "joint_extraction_results.json")
    PREDICTIONS_OUTPUT_PATH = os.path.join(output_dir, "joint_label_studio_predictions.json")
    
    print("💾 SAVING JOINT EXTRACTION RESULTS AND CREATING LABEL STUDIO FILES")
    print("=" * 70)
    
    # Save detailed processing results
    with open(JOINT_OUTPUT, 'w', encoding='utf-8') as f:
        json.dump(processing_results, f, indent=2, ensure_ascii=False)
    
    print(f"✅ Detailed results saved: {JOINT_OUTPUT}")
    
    # Create Label Studio tasks in predictions format
    label_studio_tasks = []
    
    for result_idx, result in enumerate(processing_results):
        sentence = result['sentence']
        entities = result['entities']  # Now extracted by the model
        relations = result['relations']
        
        prediction_result_items = []
        entity_id_map = {}  # Maps entity text to prediction ID
        
        print(f"\n🔍 Processing task {result_idx}:")
        print(f"  Sentence: {sentence[:100]}...")
        print(f"  Entities: {len(entities)} (extracted by model)")
        print(f"  Relations: {len(relations)}")
        
        # 1. Create entity predictions from model-extracted entities
        for i, entity in enumerate(entities):
            unique_entity_pred_id = f"joint_ent_{result_idx}_{i}"
            entity_id_map[entity['text']] = unique_entity_pred_id
            
            # Only add if position is valid
            if entity['start_char'] >= 0:
                entity_prediction = {
                    "value": {
                        "start": entity['start_char'],
                        "end": entity['end_char'],
                        "text": entity['text'],
                        "labels": [entity['label']]
                    },
                    "id": unique_entity_pred_id,
                    "from_name": "label",
                    "to_name": "text",
                    "type": "labels",
                    "score": 0.95
                }
                prediction_result_items.append(entity_prediction)
                print(f"    Entity: {entity['text']} ({entity['label']}) -> {unique_entity_pred_id}")
        
        # 2. Create relation predictions
        relations_added = 0
        relations_skipped = 0
        
        for j, relation in enumerate(relations):
            source_text = relation['source']
            target_text = relation['target']
            
            # Find matching entities
            source_entity_pred_id = entity_id_map.get(source_text)
            target_entity_pred_id = entity_id_map.get(target_text)
            
            # Try fuzzy matching if exact match fails
            if not source_entity_pred_id:
                source_entity_pred_id = find_best_entity_match(source_text, entity_id_map)
            
            if not target_entity_pred_id:
                target_entity_pred_id = find_best_entity_match(target_text, entity_id_map)
            
            if source_entity_pred_id and target_entity_pred_id:
                # Map confidence to score
                confidence_score = {
                    "HIGH": 0.9,
                    "MEDIUM": 0.7,
                    "LOW": 0.5,
                    "UNKNOWN": 0.6
                }.get(relation.get('confidence', 'UNKNOWN'), 0.6)
                
                relation_prediction = {
                    "from_id": source_entity_pred_id,
                    "to_id": target_entity_pred_id,
                    "type": "relation",
                    "direction": "right",
                    "labels": [relation['relationship']],
                    "from_name": "relation",
                    "to_name": "label",
                    "score": confidence_score,
                    "meta": {
                        "confidence": relation.get('confidence', 'UNKNOWN'),
                        "discovered_by": "joint_extraction",
                        "original_source": source_text,
                        "original_target": target_text
                    }
                }
                prediction_result_items.append(relation_prediction)
                relations_added += 1
                print(f"    ✅ Relation: {source_text} --[{relation['relationship']}]--> {target_text}")
            else:
                relations_skipped += 1
                print(f"    ❌ Skipped relation: {source_text} --[{relation['relationship']}]--> {target_text}")
        
        print(f"  Relations added: {relations_added}, Skipped: {relations_skipped}")
        
        # Create the top-level task with predictions
        formatted_task = {
            "data": {
                "text": sentence
            },
            "predictions": [{
                "model_version": "mistral-joint-extraction-v1.0",
                "score": 0.85,
                "result": prediction_result_items
            }],
            "id": result.get('sentence_id', result_idx),
            "meta": {
                "method_used": "joint_entity_relation_extraction",
                "num_entities": len(entities),
                "num_relations": len(relations),
                "num_relations_mapped": relations_added,
                "num_relations_skipped": relations_skipped
            }
        }
        
        label_studio_tasks.append(formatted_task)
    
    # Save Label Studio predictions file
    with open(PREDICTIONS_OUTPUT_PATH, 'w', encoding='utf-8') as f:
        json.dump(label_studio_tasks, f, indent=2, ensure_ascii=False)
    
    print(f"\n✅ Label Studio predictions file saved: {PREDICTIONS_OUTPUT_PATH}")
    print(f"📊 Created {len(label_studio_tasks)} Label Studio tasks")
    
    # Create dynamic Label Studio configuration
    create_dynamic_label_studio_config_fixed(processing_results, output_dir)
    
    return label_studio_tasks
```

```python
def create_zero_shot_entity_extraction_prompt(sentence):
    """Create zero-shot prompt for entity extraction only."""
    
    prompt = f"""<s>[INST] Extract all entities from this Land Use/Land Cover sentence.

**Entity Types:**
- CHANGE: Transformation words (increased, decreased, converted, expanded, reduced, grew, declined)
- LOC: Location names (countries, cities, regions, areas, zones)
- LULC: Land Use/Land Cover types (forest, cropland, urban area, built-up area, grassland, wetland, water body, bare ground, agricultural land)
- DATE: Temporal references (years, months, periods, seasons)
- PERCENT: Percentage values (25%, 10.5%, thirty percent)
- CARDINAL: Numeric values without % (1000, 2.5 million, three)
- COORDINATES: Geographic coordinates (40.7°N, latitude 23.5)
- SURFACE_UNIT: Area measurements (100 hectares, 50 km², 1000 acres)
- PROCESS: Environmental processes (deforestation, urbanization, afforestation, degradation)
- QUANTITY: Other quantities (rate of change, annual loss, total area)

**Rules:**
- Extract the exact text as it appears in the sentence
- All numbers with % are PERCENT
- All numbers without % are CARDINAL
- Be thorough and precide 
- there is no entity called  BUILT_UP_AREA or event 
- the term built up area must label as LULC
- the name of the entity must be one of the Entity Types dont halusonate and add more labels 

**Example:**
Sentence: "Forest area in Brazil decreased by 15% between 2010 and 2020 due to agricultural expansion."

ENTITIES:
- Forest area | LULC
- Brazil | LOC
- decreased | CHANGE
- 15% | PERCENT
- 2010 | DATE
- 2020 | DATE
- agricultural expansion | PROCESS

**Now analyze this sentence:**
{sentence}

Output ONLY the entities in the exact format shown above.
[/INST]"""
    
    return prompt
```

```python
def generate_entity_extraction(sentence, model, tokenizer):
    """Generate entities only using zero-shot approach."""
    
    print(f"\n📄 Processing sentence: '{sentence}'")
    print("=" * 80)
    
    # Create entity-only prompt
    prompt = create_zero_shot_entity_extraction_prompt(sentence)
    
    # Tokenize
    inputs = tokenizer(
        prompt, 
        return_tensors="pt", 
        truncation=True, 
        max_length=2048,
        padding=True
    )
    
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    
    # Generate
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=300,  # Less tokens needed for entities only
            temperature=0.1,     # Low temperature for consistency
            do_sample=False,     # Deterministic
            pad_token_id=tokenizer.eos_token_id,
            repetition_penalty=1.1
        )
    
    # Decode
    generated_text = tokenizer.decode(
        outputs[0][inputs['input_ids'].shape[1]:],
        skip_special_tokens=True
    )
    
    print(f"🤖 Model response:\n{generated_text}")
    print("=" * 80)
    
    return generated_text, prompt
```

```python
def parse_entity_response(response, original_sentence):
    """Parse entity-only model response."""
    
    entities = []
    lines = response.strip().split('\n')
    
    print(f"🔍 Parsing {len(lines)} lines...")
    
    found_entities_section = False
    
    for line_num, line in enumerate(lines):
        line = line.strip()
        
        if not line:
            continue
            
        # Check if we found the ENTITIES: header
        if 'ENTITIES:' in line.upper():
            found_entities_section = True
            print(f"  Line {line_num}: Found ENTITIES section")
            continue
        
        # Parse entity lines (with or without dash)
        if '|' in line and (found_entities_section or line.startswith('-')):
            # Remove leading dash if present
            line = line.lstrip('- ').strip()
            parts = line.split('|')
            
            if len(parts) >= 2:
                entity_text = parts[0].strip()
                entity_type = parts[1].strip()
                
                # Clean up entity type (remove extra info)
                entity_type = entity_type.split(',')[0].strip()
                entity_type = entity_type.split('(')[0].strip()
                
                # Find position in original sentence
                start_pos = original_sentence.lower().find(entity_text.lower())
                end_pos = start_pos + len(entity_text) if start_pos != -1 else -1
                
                entity = {
                    "text": entity_text,
                    "label": entity_type,
                    "start_char": start_pos,
                    "end_char": end_pos
                }
                
                entities.append(entity)
                print(f"  ✅ Entity found: {entity_text} ({entity_type}) [{start_pos}:{end_pos}]")
            else:
                print(f"  ❌ Could not parse line: '{line}'")
    
    print(f"📊 Total entities parsed: {len(entities)}")
    return entities
```

```python
def parse_entity_response(response, original_sentence):
    """Parse entity-only model response."""
    
    entities = []
    lines = response.strip().split('\n')
    
    print(f"🔍 Parsing {len(lines)} lines...")
    
    found_entities_section = False
    
    for line_num, line in enumerate(lines):
        line = line.strip()
        
        if not line:
            continue
            
        # Check if we found the ENTITIES: header
        if 'ENTITIES:' in line.upper():
            found_entities_section = True
            print(f"  Line {line_num}: Found ENTITIES section")
            continue
        
        # Parse entity lines (with or without dash)
        if '|' in line and (found_entities_section or line.startswith('-')):
            # Remove leading dash if present
            line = line.lstrip('- ').strip()
            parts = line.split('|')
            
            if len(parts) >= 2:
                entity_text = parts[0].strip()
                entity_type = parts[1].strip()
                
                # Clean up entity type (remove extra info)
                entity_type = entity_type.split(',')[0].strip()
                entity_type = entity_type.split('(')[0].strip()
                
                # Find position in original sentence
                start_pos = original_sentence.lower().find(entity_text.lower())
                end_pos = start_pos + len(entity_text) if start_pos != -1 else -1
                
                entity = {
                    "text": entity_text,
                    "label": entity_type,
                    "start_char": start_pos,
                    "end_char": end_pos
                }
                
                entities.append(entity)
                print(f"  ✅ Entity found: {entity_text} ({entity_type}) [{start_pos}:{end_pos}]")
            else:
                print(f"  ❌ Could not parse line: '{line}'")
    
    print(f"📊 Total entities parsed: {len(entities)}")
    return entities
```

```python
def process_sentences_entity_only(sentences_data, model, tokenizer, max_sentences=None):
    """Process sentences for entity extraction only."""
    
    all_results = []
    
    # Extract sentences
    if isinstance(sentences_data, list) and len(sentences_data) > 0:
        if 'original_sentence' in sentences_data[0]:
            sentences = [item['original_sentence'] for item in sentences_data]
        else:
            sentences = sentences_data
    else:
        sentences = sentences_data
    
    if max_sentences:
        sentences = sentences[:max_sentences]
    
    print(f"🚀 Starting ENTITY-ONLY extraction for {len(sentences)} sentences...")
    
    for i, sentence in enumerate(tqdm(sentences, desc="Entity extraction")):
        try:
            # Generate extraction
            response, prompt = generate_entity_extraction(sentence, model, tokenizer)
            
            # Parse response
            entities = parse_entity_response(response, sentence)
            
            # Store results
            result = {
                "sentence_id": i,
                "sentence": sentence,
                "entities": entities,
                "raw_response": response,
                "method": "zero-shot-entities"
            }
            
            all_results.append(result)
            
            print(f"\n✅ Sentence {i+1}: Found {len(entities)} entities")
            
        except Exception as e:
            print(f"❌ Error processing sentence {i}: {e}")
            continue
    
    return all_results
```

```python
def save_entities_to_label_studio(results, output_dir="entity_output"):
    """Save entity-only results in Label Studio format."""
    
    import os
    import json
    import xml.sax.saxutils
    
    os.makedirs(output_dir, exist_ok=True)
    
    label_studio_tasks = []
    all_entity_types = set()
    
    # Process each result
    for idx, result in enumerate(results):
        sentence = result['sentence']
        entities = result['entities']
        
        prediction_result_items = []
        
        # Collect entity types
        for entity in entities:
            all_entity_types.add(entity['label'])
        
        # Create entity predictions
        for i, entity in enumerate(entities):
            # Only add if position is valid
            if entity['start_char'] >= 0:
                entity_prediction = {
                    "value": {
                        "start": entity['start_char'],
                        "end": entity['end_char'],
                        "text": entity['text'],
                        "labels": [entity['label']]
                    },
                    "id": f"entity_{idx}_{i}",
                    "from_name": "label",
                    "to_name": "text",
                    "type": "labels",
                    "score": 0.9
                }
                prediction_result_items.append(entity_prediction)
        
        # Create Label Studio task
        task = {
            "data": {
                "text": sentence,
                "meta": {
                    "sentence_id": idx,
                    "extraction_method": "zero-shot-entities"
                }
            },
            "predictions": [{
                "model_version": "mistral-entity-extraction-v1.0",
                "score": 0.9,
                "result": prediction_result_items
            }],
            "id": idx,
            "meta": {
                "num_entities": len(entities)
            }
        }
        
        label_studio_tasks.append(task)
    
    # Save Label Studio JSON
    label_studio_path = os.path.join(output_dir, "entity_label_studio.json")
    with open(label_studio_path, 'w', encoding='utf-8') as f:
        json.dump(label_studio_tasks, f, indent=2, ensure_ascii=False)
    
    print(f"✅ Saved Label Studio format to: {label_studio_path}")
    
    # Create Label Studio XML config for entities only
    create_entity_label_studio_config(all_entity_types, output_dir)
    
    # Save raw results
    raw_results_path = os.path.join(output_dir, "entity_extraction_results.json")
    with open(raw_results_path, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)
    
    print(f"✅ Saved raw results to: {raw_results_path}")
    
    return label_studio_tasks

def create_entity_label_studio_config(entity_types, output_dir):
    """Create Label Studio configuration XML for entities only."""
    
    import xml.sax.saxutils
    
    colors = [
        "#FF6B6B", "#4ECDC4", "#45B7D1", "#96CEB4", "#FECA57", 
        "#FF9FF3", "#A55EEA", "#54A0FF", "#5F27CD", "#00D2D3"
    ]
    
    # Build entity labels
    entity_labels = ""
    for i, entity_type in enumerate(sorted(entity_types)):
        color = colors[i % len(colors)]
        safe_entity = xml.sax.saxutils.escape(entity_type)
        entity_labels += f'    <Label value="{safe_entity}" background="{color}"/>\n'
    
    # Create config XML
    config_xml = f"""<View>
  <Text name="text" value="$text"/>
  
  <Labels name="label" toName="text">
{entity_labels}  </Labels>
</View>"""
    
    # Save config
    config_path = os.path.join(output_dir, "entity_label_studio_config.xml")
    with open(config_path, 'w', encoding='utf-8') as f:
        f.write(config_xml)
    
    print(f"✅ Saved Label Studio config to: {config_path}")
    print(f"📋 Configured {len(entity_types)} entity types")
```
