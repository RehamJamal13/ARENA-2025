```python
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification
import json
from tqdm import tqdm

def extract_entities_from_text(text, model, tokenizer, device):
    """
    Extract entities from a single text using the trained model.
    """
    model.eval()
    
    # Tokenize input
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512,
        return_offsets_mapping=True,
        padding=True
    )
    
    # Move inputs to device
    input_ids = inputs['input_ids'].to(device)
    attention_mask = inputs['attention_mask'].to(device)
    
    # Get predictions
    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        predictions = torch.argmax(outputs.logits, dim=-1)
    
    # Convert predictions to labels
    tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
    labels = [model.config.id2label[pred.item()] for pred in predictions[0]]
    offset_mapping = inputs['offset_mapping'][0]
    
    # Extract entities
    entities = []
    current_entity = None
    
    for idx, (token, label, offset) in enumerate(zip(tokens, labels, offset_mapping)):
        if token in tokenizer.all_special_tokens:
            continue
            
        if label.startswith('B-'):
            # Start of new entity
            if current_entity:
                entities.append(current_entity)
            current_entity = {
                'text': tokenizer.convert_tokens_to_string([token]),
                'label': label[2:],
                'start': offset[0].item(),
                'end': offset[1].item(),
                'tokens': [token]
            }
        elif label.startswith('I-') and current_entity and label[2:] == current_entity['label']:
            # Continuation of current entity
            current_entity['tokens'].append(token)
            current_entity['end'] = offset[1].item()
            current_entity['text'] = tokenizer.convert_tokens_to_string(current_entity['tokens'])
        else:
            # Not an entity
            if current_entity:
                entities.append(current_entity)
                current_entity = None
    
    if current_entity:
        entities.append(current_entity)
    
    # Clean up entity text
    for entity in entities:
        entity['text'] = entity['text'].strip()
        del entity['tokens']
    
    return entities
```

```python
def process_csv_to_json(csv_path, model_path, output_json_path):
    """
    Process a CSV file and extract entities, saving results to JSON.
    """
    # Load model and tokenizer
    print("Loading model and tokenizer...")
    tokenizer = AutoTokenizer.from_pretrained(model_path)
    model = AutoModelForTokenClassification.from_pretrained(model_path)
    
    # Set device
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)
    print(f"Using device: {device}")
    
    # Load CSV file
    print(f"Loading CSV file: {csv_path}")
    df = pd.read_csv(csv_path)
    
    # Check available columns
    print(f"Available columns: {list(df.columns)}")
    
    # Define the text column based on what's available
    if 'processed_text' in df.columns:
        text_column = 'processed_text'
    elif 'text_segment' in df.columns:
        text_column = 'text_segment'
    elif 'text' in df.columns:
        text_column = 'text'
    else:
        # Use the first column if nothing matches
        text_column = df.columns[0]
    
    print(f"Using text column: '{text_column}'")
    
    results = []
    
    # Process each text
    print("Extracting entities...")
    for idx, row in tqdm(df.iterrows(), total=len(df)):
        text = str(row[text_column])  # Use the correct column name
        
        # Extract entities
        entities = extract_entities_from_text(text, model, tokenizer, device)
        
        # Create result with ONLY sentence and entities
        result = {
            'sentence': text,
            'entities': entities
        }
        
        results.append(result)
    
    # Save to JSON
    print(f"Saving results to: {output_json_path}")
    with open(output_json_path, 'w', encoding='utf-8') as f:
        json.dump(results, f, indent=2, ensure_ascii=False)
    
    # Print summary
    total_entities = sum(len(r['entities']) for r in results)
    print(f"\nProcessing complete!")
    print(f"Total sentences processed: {len(results)}")
    print(f"Total entities extracted: {total_entities}")
    
    return results
```

```python
import pandas as pd
import re

# File paths and column names
CSV_FILE_PATH = 'annotated_corpus_for_dataverse.csv'
PROCESSED_CSV_PATH = 'processed_corpus_for_ner.csv'

# Column names
ID_COLUMN = 'id_segment'
TEXT_COLUMN = 'text_segment'
RELEVANCE_COLUMN = 'relevance_label'

import re
import json

import json
import re

def preprocess_tei_json(text):
    if not text or text.strip() == '':
        return ''

    try:
        # Try to parse as JSON
        data = json.loads(text.replace("'", '"'))
        
        # Extract text content
        text_parts = []
        
        if 'head' in data and '#text' in data['head']:
            text_parts.append(data['head']['#text'].replace('[', '').replace(']', ''))
        
        if 'p' in data and '#text' in data['p']:
            text_parts.append(data['p']['#text'])
        
        # Join the parts
        text = '. '.join(text_parts)
        
    except:
        # If parsing fails, fall back to regex cleaning
        text = re.sub(r"'@[^']*':\s*'[^']*',?\s*", '', text)  # Remove attributes
        text = re.sub(r"'#text':\s*'", '', text)  # Remove #text markers
        text = re.sub(r"[{}'\[\]]", '', text)  # Remove JSON punctuation
        
        # Remove URLs
        text = re.sub(r'http[s]?://[^\s,]+', '', text)
        
        # Remove brackets content
        text = re.sub(r'\[.*?\]', '', text)
        
        # Clean up punctuation and spacing
        text = re.sub(r'\s+', ' ', text)
        text = re.sub(r'\s*,\s*,\s*', ', ', text)
        text = re.sub(r'^\s*[,:]\s*', '', text)
        text = re.sub(r'\s*[,:]\s*$', '', text)
    
    return text.strip()

# Read the CSV file
print(f"📂 Reading CSV file: {CSV_FILE_PATH}")
df = pd.read_csv(CSV_FILE_PATH)
print(f"✅ Loaded {len(df)} rows from CSV")

# Display column info
print(f"\n📊 Available columns: {list(df.columns)}")
print(f"📝 Using text column: '{TEXT_COLUMN}'")
print(f"🏷️  Using ID column: '{ID_COLUMN}'")

# Check if relevance column exists
if RELEVANCE_COLUMN in df.columns:
    # Filter for rows where relevance_label is not 4
    sentences_to_process_df = df[df[RELEVANCE_COLUMN] == 2].copy()
    print(f"\n✅ Filtered to {len(sentences_to_process_df)} rows with {RELEVANCE_COLUMN} != 4")
    print(f"❌ Excluded {len(df) - len(sentences_to_process_df)} rows with {RELEVANCE_COLUMN} = 4")
else:
    print(f"\n⚠️  Column '{RELEVANCE_COLUMN}' not found. Processing all rows.")
    sentences_to_process_df = df.copy()

# Show some examples of problematic texts before preprocessing
print(f"\n🔍 Examples of texts that need cleaning:")
problematic_patterns = ['xmlns:', 'http:', 'head:', 'text:', 'ref:', '<', '>']
for pattern in problematic_patterns:
    sample = sentences_to_process_df[sentences_to_process_df[TEXT_COLUMN].astype(str).str.contains(pattern, na=False)].head(1)
    if not sample.empty:
        print(f"\nExample with '{pattern}':")
        print(f"Original: {sample.iloc[0][TEXT_COLUMN][:200]}...")
        break

# Apply preprocessing
print(f"\n🔧 Preprocessing text...")
sentences_to_process_df['processed_text'] = sentences_to_process_df[TEXT_COLUMN].apply(preprocess_tei_json)

# Remove empty texts after preprocessing
before_empty_removal = len(sentences_to_process_df)
sentences_to_process_df = sentences_to_process_df[sentences_to_process_df['processed_text'].str.len() > 0]
print(f"✅ Removed {before_empty_removal - len(sentences_to_process_df)} empty texts after preprocessing")

# Display sample of preprocessed data
print(f"\n📋 Sample of preprocessed data:")
sample_size = min(5, len(sentences_to_process_df))
for idx in range(sample_size):
    row = sentences_to_process_df.iloc[idx]
    print(f"\n--- Sample {idx + 1} ---")
    print(f"ID: {row[ID_COLUMN]}")
    original_text = str(row[TEXT_COLUMN])[:150]
    processed_text = str(row['processed_text'])[:150]
    print(f"Original: {original_text}...")
    print(f"Processed: {processed_text}...")

# Test the specific problematic example
test_text = ")., xmlns: http:www.tei-c.orgns1.0, head: n: 4.2., text: change in the study site, p: ref: , text: From 1996 to 2017, the area covered by cropped fields, C, has increased from 40% to 51.3% over the study site area, and from 47% to 64% of the arable lands in the study site due to the loss fallows"
print(f"\n🧪 Testing problematic example:")
print(f"Original: {test_text}")
print(f"Cleaned: {preprocess_tei_json(test_text)}")

# Save processed data
sentences_to_process_df.to_csv(PROCESSED_CSV_PATH, index=False)
print(f"\n💾 Saved processed data to: {PROCESSED_CSV_PATH}")
print(f"✅ Total rows ready for NER: {len(sentences_to_process_df)}")

# Show statistics
print(f"\n📊 Text length statistics:")
text_lengths = sentences_to_process_df['processed_text'].str.len()
print(f"  - Mean length: {text_lengths.mean():.1f} characters")
print(f"  - Min length: {text_lengths.min()} characters")
print(f"  - Max length: {text_lengths.max()} characters")
print(f"  - Texts > 512 chars: {(text_lengths > 512).sum()} (may be truncated by model)")
```
