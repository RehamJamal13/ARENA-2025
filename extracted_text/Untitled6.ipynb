```python
import json

def merge_json_files(file1_path, file2_path, output_file_path):
    """
    Merges two JSON files into a single JSON file.

    Args:
        file1_path (str): The path to the first JSON file.
        file2_path (str): The path to the second JSON file.
        output_file_path (str): The path to the output JSON file.
    """
    merged_data = []

    try:
        # Load data from the first file
        with open(file1_path, 'r') as f1:
            data1 = json.load(f1)
            merged_data.extend(data1)
    except FileNotFoundError:
        print(f"Error: File not found: {file1_path}")
        return
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in: {file1_path}")
        return

    try:
        # Load data from the second file
        with open(file2_path, 'r') as f2:
            data2 = json.load(f2)
            merged_data.extend(data2)
    except FileNotFoundError:
        print(f"Error: File not found: {file2_path}")
        return
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in: {file2_path}")
        return

    # Write the merged data to the output file
    try:
        with open(output_file_path, 'w') as outfile:
            json.dump(merged_data, outfile, indent=4)  # Use indent for readability        print(f"Successfully merged data into: {output_file_path}")
    except IOError:
        print(f"Error: Could not write to file: {output_file_path}")
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in: {output_file_path}")
        return

# Example usage (replace with your file paths):
file1_path = 'extracted_ALL_entities_structured.json'  # Replace with the path to your first JSON file
file2_path = 'extracted_ALL_entities_structured2.json'  # Replace with the path to your second JSON file
output_file_path = 'merged_file.json'  # Replace with the desired output file path

merge_json_files(file1_path, file2_path, output_file_path)
```
