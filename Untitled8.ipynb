```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json


def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]
    
    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence (RELATIONS ONLY)"""
    
    df = pd.read_csv(csv_path)
    
    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")
    
    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])
    
    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/SOURCE')
    if not target_col:
        missing_cols.append('target/TARGET')
    if not relation_col:
        missing_cols.append('relationship/relation/RELATION')
    if not sentence_id_col:
        missing_cols.append('sentence_id/SENTENCE_ID/id')
    if not sentence_col:
        missing_cols.append('sentence/text/TEXT_SEGMENT')
    
    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")
    
    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")
    
    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row[sentence_id_col]
        sentence_text = row[sentence_col]
        
        # Skip empty rows (sentences with no relations)

        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract relations (source, relation, target) - ONLY RELATIONS
        relation = (str(row[source_col]).strip(), 
                   str(row[relation_col]).strip(), 
                   str(row[target_col]).strip())
        sentence_data[sentence_id]['relations'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_relation_errors(gt_relations: Set, pred_relations: Set) -> Dict:
    """Analyze relation-level errors (source, relation, target)"""
    
    errors = {
        'false_positives': list(pred_relations - gt_relations),
        'false_negatives': list(gt_relations - pred_relations),
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Group errors by relation type (middle element of the tuple)
    for relation in errors['false_positives']:
        relation_type = relation[1]  # relation type is at index 1
        errors['by_type'][relation_type]['fp'].append(relation)
        
    for relation in errors['false_negatives']:
        relation_type = relation[1]  # relation type is at index 1
        errors['by_type'][relation_type]['fn'].append(relation)
    
    return dict(errors)

def evaluate_relations_only(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files (RELATIONS ONLY)"""
    
    print("📂 Loading data...")
    gt_data = load_csv_relations_only(ground_truth_csv)
    pred_data = load_csv_relations_only(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results - RELATIONS ONLY
    results = {
        'overall_relations': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': []
        }
    }
    
    # Collect all relations
    all_gt_relations = set()
    all_pred_relations = set()
    
    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {'relations': set()})
        pred_sentence = pred_data.get(sentence_id, {'relations': set()})
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
        
        # Relation evaluation for this sentence
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall relation metrics...")
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall_relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn,
        'total_gt_relations': len(all_gt_relations),
        'total_pred_relations': len(all_pred_relations)
    }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis
    results['error_analysis']['relation_errors'] = analyze_relation_errors(all_gt_relations, all_pred_relations)
    
    return results

def print_relations_report(results: Dict):
    """Print RELATIONS ONLY evaluation report"""
    
    print("\n" + "="*70)
    print("🎯 RELATION EXTRACTION EVALUATION REPORT")
    print("="*70)
    
    # Overall relation performance
    overall = results['overall_relations']
    print(f"\n📈 OVERALL RELATION PERFORMANCE:")
    print(f"   Precision: {overall['precision']:.3f}")
    print(f"   Recall:    {overall['recall']:.3f}")
    print(f"   F1-Score:  {overall['f1']:.3f}")
    print(f"   Total GT Relations:   {overall['total_gt_relations']}")
    print(f"   Total Pred Relations: {overall['total_pred_relations']}")
    print(f"   True Positives:  {overall['tp']}")
    print(f"   False Positives: {overall['fp']}")
    print(f"   False Negatives: {overall['fn']}")
    
    # Relation performance by type
    print(f"\n🔗 RELATION PERFORMANCE BY TYPE:")
    print(f"{'Relation Type':<25} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 75)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by F1 score (descending)
    relation_items.sort(key=lambda x: x[1]['f1'], reverse=True)
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<25} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<25} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    print(f"   Relation false positives: {len(errors['relation_errors']['false_positives'])}")
    print(f"   Relation false negatives: {len(errors['relation_errors']['false_negatives'])}")
    
    # Show detailed relation errors
    print(f"\n🔍 DETAILED RELATION ERRORS:")
    relation_errors = errors['relation_errors']
    
    if relation_errors['false_positives']:
        print(f"\n   FALSE POSITIVES (Predicted but not in Ground Truth):")
        for i, rel in enumerate(relation_errors['false_positives'][:10]):  # Show first 10
            print(f"      {i+1}. {rel}")
        if len(relation_errors['false_positives']) > 10:
            print(f"      ... and {len(relation_errors['false_positives'])-10} more")
    
    if relation_errors['false_negatives']:
        print(f"\n   FALSE NEGATIVES (In Ground Truth but not Predicted):")
        for i, rel in enumerate(relation_errors['false_negatives'][:10]):  # Show first 10
            print(f"      {i+1}. {rel}")
        if len(relation_errors['false_negatives']) > 10:
            print(f"      ... and {len(relation_errors['false_negatives'])-10} more")
    
    # Show errors by relation type
    print(f"\n📊 ERRORS BY RELATION TYPE:")
    for rel_type, type_errors in relation_errors['by_type'].items():
        fp_count = len(type_errors['fp'])
        fn_count = len(type_errors['fn'])
        if fp_count > 0 or fn_count > 0:
            print(f"   {rel_type}: FP={fp_count}, FN={fn_count}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "LabelledRelations_Increase_Decrease - Feuille 1.csv"
PREDICTIONS_CSV = "relations_table_20250709_165543 - GEMENI AI PRO 2.5 SMALL SET ZEROSHOT.csv"

if __name__ == "__main__":
    try:
        # Run RELATIONS ONLY evaluation
        results = evaluate_relations_only(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print detailed report
        print_relations_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "relation_evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json
def normalize_relation_type(relation_type: str) -> str:
    """Normalize relation types to handle case and tense variations"""
    # Convert to uppercase
    normalized = relation_type.upper()
    
    # Handle tense variations (you can customize these rules)
    tense_mappings = {
        'INCREASED_BY': 'INCREASES_BY',
        'DECREASED_BY': 'DECREASES_BY',
        'INCREASED': 'INCREASE',
        'DECREASED': 'DECREASE',
        'DECREASES_by' : 'DECREASES_BY'
        
        
    }
    
    if normalized in tense_mappings:
        normalized = tense_mappings[normalized]
    
    return normalized


def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]
    
    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence (RELATIONS ONLY)"""
    
    df = pd.read_csv(csv_path)
    
    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")
    
    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])
    
    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/SOURCE')
    if not target_col:
        missing_cols.append('target/TARGET')
    if not relation_col:
        missing_cols.append('relationship/relation/RELATION')
    if not sentence_id_col:
        missing_cols.append('sentence_id/SENTENCE_ID/id')
    if not sentence_col:
        missing_cols.append('sentence/text/TEXT_SEGMENT')
    
    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")
    
    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")
    
    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {'relations': set(), 'relation_to_sentence': {}})
    
    for _, row in df.iterrows():
        sentence_id = row[sentence_id_col]
        sentence_text = row[sentence_col]
        
        # Skip empty rows (sentences with no relations)
        
        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract relations (source, relation, target) - ONLY RELATIONS
        relation = (str(row[source_col]).strip(), 
                   str(row[relation_col]).strip(), 
                   str(row[target_col]).strip())
        sentence_data[sentence_id]['relations'].add(relation)
        
        # Map relation to sentence for error reporting
        sentence_data[sentence_id]['relation_to_sentence'][relation] = sentence_text
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_relation_errors(gt_data: Dict, pred_data: Dict) -> Dict:
    """Analyze relation-level errors with sentence information"""
    
    errors = {
        'false_positives': [],
        'false_negatives': [],
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Collect all relations with their sentences
    gt_relations_with_sentences = {}
    pred_relations_with_sentences = {}
    
    for sentence_id, data in gt_data.items():
        for relation in data['relations']:
            gt_relations_with_sentences[relation] = data.get('sentence', f'Sentence ID: {sentence_id}')
    
    for sentence_id, data in pred_data.items():
        for relation in data['relations']:
            pred_relations_with_sentences[relation] = data.get('sentence', f'Sentence ID: {sentence_id}')
    
    # Find false positives (in pred but not in gt)
    for relation, sentence in pred_relations_with_sentences.items():
        if relation not in gt_relations_with_sentences:
            errors['false_positives'].append({
                'relation': relation,
                'sentence': sentence
            })
            relation_type = relation[1]
            errors['by_type'][relation_type]['fp'].append({
                'relation': relation,
                'sentence': sentence
            })
    
    # Find false negatives (in gt but not in pred)
    for relation, sentence in gt_relations_with_sentences.items():
        if relation not in pred_relations_with_sentences:
            errors['false_negatives'].append({
                'relation': relation,
                'sentence': sentence
            })
            relation_type = relation[1]
            errors['by_type'][relation_type]['fn'].append({
                'relation': relation,
                'sentence': sentence
            })
    
    return dict(errors)

def evaluate_relations_only(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files (RELATIONS ONLY)"""
    
    print("📂 Loading data...")
    gt_data = load_csv_relations_only(ground_truth_csv)
    pred_data = load_csv_relations_only(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results - RELATIONS ONLY
    results = {
        'overall_relations': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': []
        }
    }
    
    # Collect all relations
    all_gt_relations = set()
    all_pred_relations = set()
    
    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {'relations': set()})
        pred_sentence = pred_data.get(sentence_id, {'relations': set()})
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
        
        # Relation evaluation for this sentence
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall relation metrics...")
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall_relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn,
        'total_gt_relations': len(all_gt_relations),
        'total_pred_relations': len(all_pred_relations)
    }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis with sentences
    results['error_analysis']['relation_errors'] = analyze_relation_errors(gt_data, pred_data)
    
    return results

def print_relations_report(results: Dict):
    """Print RELATIONS ONLY evaluation report"""
    
    print("\n" + "="*70)
    print("🎯 RELATION EXTRACTION EVALUATION REPORT")
    print("="*70)
    
    # Overall relation performance
    overall = results['overall_relations']
    print(f"\n📈 OVERALL RELATION PERFORMANCE:")
    print(f"   Precision: {overall['precision']:.3f}")
    print(f"   Recall:    {overall['recall']:.3f}")
    print(f"   F1-Score:  {overall['f1']:.3f}")
    print(f"   Total GT Relations:   {overall['total_gt_relations']}")
    print(f"   Total Pred Relations: {overall['total_pred_relations']}")
    print(f"   True Positives:  {overall['tp']}")
    print(f"   False Positives: {overall['fp']}")
    print(f"   False Negatives: {overall['fn']}")
    
    # Relation performance by type
    print(f"\n🔗 RELATION PERFORMANCE BY TYPE:")
    print(f"{'Relation Type':<25} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 75)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by F1 score (descending)
    relation_items.sort(key=lambda x: x[1]['f1'], reverse=True)
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<25} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<25} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    print(f"   Relation false positives: {len(errors['relation_errors']['false_positives'])}")
    print(f"   Relation false negatives: {len(errors['relation_errors']['false_negatives'])}")
    
    # Show detailed relation errors with sentences
    print(f"\n🔍 DETAILED RELATION ERRORS:")
    relation_errors = errors['relation_errors']
    
    if relation_errors['false_positives']:
        print(f"\n   FALSE POSITIVES (Predicted but not in Ground Truth):")
        for i, error_info in enumerate(relation_errors['false_positives'][:100]):  # Show first 10
            rel = error_info['relation']
            sentence = error_info['sentence']
            print(f"      {i+1}. Relation: {rel}")
            print(f"         Sentence: {sentence[:100]}{'...' if len(sentence) > 100 else ''}")
            print()
        if len(relation_errors['false_positives']) > 10:
            print(f"      ... and {len(relation_errors['false_positives'])-100} more")
    
    if relation_errors['false_negatives']:
        print(f"\n   FALSE NEGATIVES (In Ground Truth but not Predicted):")
        for i, error_info in enumerate(relation_errors['false_negatives'][:100]):  # Show first 10
            rel = error_info['relation']
            sentence = error_info['sentence']
            print(f"      {i+1}. Relation: {rel}")
            print(f"         Sentence: {sentence[:100]}{'...' if len(sentence) > 100 else ''}")
            print()
        if len(relation_errors['false_negatives']) > 10:
            print(f"      ... and {len(relation_errors['false_negatives'])-100} more")
    
    # Show errors by relation type
    print(f"\n📊 ERRORS BY RELATION TYPE:")
    for rel_type, type_errors in relation_errors['by_type'].items():
        fp_count = len(type_errors['fp'])
        fn_count = len(type_errors['fn'])
        if fp_count > 0 or fn_count > 0:
            print(f"   {rel_type}: FP={fp_count}, FN={fn_count}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "LabelledRelations_Increase_Decrease - Feuille 1.csv"
PREDICTIONS_CSV = "relations_table_20250709_165543 - claude ai pipline approtch (1).csv"

if __name__ == "__main__":
    try:
        # Run RELATIONS ONLY evaluation
        results = evaluate_relations_only(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print detailed report
        print_relations_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "relation_evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json


def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]
    
    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence (RELATIONS ONLY)"""
    
    df = pd.read_csv(csv_path)
    
    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")
    
    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])
    
    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/SOURCE')
    if not target_col:
        missing_cols.append('target/TARGET')
    if not relation_col:
        missing_cols.append('relationship/relation/RELATION')
    if not sentence_id_col:
        missing_cols.append('sentence_id/SENTENCE_ID/id')
    if not sentence_col:
        missing_cols.append('sentence/text/TEXT_SEGMENT')
    
    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")
    
    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")
    
    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {
        'relations': set(),
        'sources': set(),
        'targets': set(), 
        'relation_types': set()
    })
    
    for _, row in df.iterrows():
        sentence_id = row[sentence_id_col]
        sentence_text = row[sentence_col]
        source = str(row[source_col]).strip()
        target = str(row[target_col]).strip()
        relation = str(row[relation_col]).strip()

        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract full relations (source, relation, target)
        full_relation = (source, relation, target)
        sentence_data[sentence_id]['relations'].add(full_relation)
        
        # Extract individual components
        sentence_data[sentence_id]['sources'].add(source)
        sentence_data[sentence_id]['targets'].add(target)
        sentence_data[sentence_id]['relation_types'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_component_errors(gt_components: Set, pred_components: Set, component_name: str) -> Dict:
    """Analyze component-level errors"""
    
    errors = {
        'false_positives': list(pred_components - gt_components),
        'false_negatives': list(gt_components - pred_components),
        'component_name': component_name
    }
    
    return errors

def evaluate_individual_components(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files with individual component analysis"""
    
    print("📂 Loading data...")
    gt_data = load_csv_relations_only(ground_truth_csv)
    pred_data = load_csv_relations_only(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results
    results = {
        'overall_relations': {},
        'overall_sources': {},
        'overall_targets': {},
        'overall_relation_types': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': {},
            'source_errors': {},
            'target_errors': {},
            'relation_type_errors': {}
        }
    }
    
    # Collect all components
    all_gt_relations = set()
    all_pred_relations = set()
    all_gt_sources = set()
    all_pred_sources = set()
    all_gt_targets = set()
    all_pred_targets = set()
    all_gt_relation_types = set()
    all_pred_relation_types = set()
    
    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        pred_sentence = pred_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
        
        # Evaluation for this sentence - FULL RELATIONS
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        # Evaluation for this sentence - SOURCES ONLY
        gt_sources = gt_sentence['sources']
        pred_sources = pred_sentence['sources']
        source_p, source_r, source_f1, source_tp, source_fp, source_fn = calculate_metrics(gt_sources, pred_sources)
        sentence_result.update({
            'source_precision': source_p,
            'source_recall': source_r,
            'source_f1': source_f1,
            'source_tp': source_tp,
            'source_fp': source_fp,
            'source_fn': source_fn
        })
        
        # Evaluation for this sentence - TARGETS ONLY
        gt_targets = gt_sentence['targets']
        pred_targets = pred_sentence['targets']
        target_p, target_r, target_f1, target_tp, target_fp, target_fn = calculate_metrics(gt_targets, pred_targets)
        sentence_result.update({
            'target_precision': target_p,
            'target_recall': target_r,
            'target_f1': target_f1,
            'target_tp': target_tp,
            'target_fp': target_fp,
            'target_fn': target_fn
        })
        
        # Evaluation for this sentence - RELATION TYPES ONLY
        gt_rel_types = gt_sentence['relation_types']
        pred_rel_types = pred_sentence['relation_types']
        rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn = calculate_metrics(gt_rel_types, pred_rel_types)
        sentence_result.update({
            'relation_type_precision': rel_type_p,
            'relation_type_recall': rel_type_r,
            'relation_type_f1': rel_type_f1,
            'relation_type_tp': rel_type_tp,
            'relation_type_fp': rel_type_fp,
            'relation_type_fn': rel_type_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        all_gt_sources.update(gt_sources)
        all_pred_sources.update(pred_sources)
        all_gt_targets.update(gt_targets)
        all_pred_targets.update(pred_targets)
        all_gt_relation_types.update(gt_rel_types)
        all_pred_relation_types.update(pred_rel_types)
        
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall metrics...")
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall_relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn,
        'total_gt': len(all_gt_relations),
        'total_pred': len(all_pred_relations)
    }
    
    # Overall source metrics
    source_p, source_r, source_f1, source_tp, source_fp, source_fn = calculate_metrics(all_gt_sources, all_pred_sources)
    results['overall_sources'] = {
        'precision': source_p, 'recall': source_r, 'f1': source_f1,
        'tp': source_tp, 'fp': source_fp, 'fn': source_fn,
        'total_gt': len(all_gt_sources),
        'total_pred': len(all_pred_sources)
    }
    
    # Overall target metrics
    target_p, target_r, target_f1, target_tp, target_fp, target_fn = calculate_metrics(all_gt_targets, all_pred_targets)
    results['overall_targets'] = {
        'precision': target_p, 'recall': target_r, 'f1': target_f1,
        'tp': target_tp, 'fp': target_fp, 'fn': target_fn,
        'total_gt': len(all_gt_targets),
        'total_pred': len(all_pred_targets)
    }
    
    # Overall relation type metrics
    rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn = calculate_metrics(all_gt_relation_types, all_pred_relation_types)
    results['overall_relation_types'] = {
        'precision': rel_type_p, 'recall': rel_type_r, 'f1': rel_type_f1,
        'tp': rel_type_tp, 'fp': rel_type_fp, 'fn': rel_type_fn,
        'total_gt': len(all_gt_relation_types),
        'total_pred': len(all_pred_relation_types)
    }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis for all components
    results['error_analysis']['relation_errors'] = analyze_component_errors(all_gt_relations, all_pred_relations, 'relations')
    results['error_analysis']['source_errors'] = analyze_component_errors(all_gt_sources, all_pred_sources, 'sources')
    results['error_analysis']['target_errors'] = analyze_component_errors(all_gt_targets, all_pred_targets, 'targets')
    results['error_analysis']['relation_type_errors'] = analyze_component_errors(all_gt_relation_types, all_pred_relation_types, 'relation_types')
    
    return results

def print_component_evaluation_report(results: Dict):
    """Print comprehensive evaluation report with individual components"""
    
    print("\n" + "="*80)
    print("🎯 COMPREHENSIVE RELATION EXTRACTION EVALUATION REPORT")
    print("="*80)
    
    # Overall performance for all components
    components = [
        ('FULL RELATIONS', results['overall_relations']),
        ('SOURCES ONLY', results['overall_sources']),
        ('TARGETS ONLY', results['overall_targets']),
        ('RELATION TYPES ONLY', results['overall_relation_types'])
    ]
    
    for component_name, metrics in components:
        print(f"\n📈 {component_name} PERFORMANCE:")
        print(f"   Precision: {metrics['precision']:.3f}")
        print(f"   Recall:    {metrics['recall']:.3f}")
        print(f"   F1-Score:  {metrics['f1']:.3f}")
        print(f"   Total GT:   {metrics['total_gt']}")
        print(f"   Total Pred: {metrics['total_pred']}")
        print(f"   True Positives:  {metrics['tp']}")
        print(f"   False Positives: {metrics['fp']}")
        print(f"   False Negatives: {metrics['fn']}")
    
    # Relation performance by type
    print(f"\n🔗 FULL RELATION PERFORMANCE BY TYPE:")
    print(f"{'Relation Type':<25} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 75)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by F1 score (descending)
    relation_items.sort(key=lambda x: x[1]['f1'], reverse=True)
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<25} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<25} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    
    error_components = [
        ('Full Relations', errors['relation_errors']),
        ('Sources', errors['source_errors']),
        ('Targets', errors['target_errors']),
        ('Relation Types', errors['relation_type_errors'])
    ]
    
    for comp_name, comp_errors in error_components:
        print(f"   {comp_name} - FP: {len(comp_errors['false_positives'])}, FN: {len(comp_errors['false_negatives'])}")
    
    # Show detailed errors for each component
    for comp_name, comp_errors in error_components:
        print(f"\n🔍 DETAILED {comp_name.upper()} ERRORS:")
        
        if comp_errors['false_positives']:
            print(f"\n   FALSE POSITIVES (Predicted but not in Ground Truth):")
            for i, item in enumerate(comp_errors['false_positives'][:10]):  # Show first 10
                print(f"      {i+1}. {item}")
            if len(comp_errors['false_positives']) > 10:
                print(f"      ... and {len(comp_errors['false_positives'])-10} more")
        
        if comp_errors['false_negatives']:
            print(f"\n   FALSE NEGATIVES (In Ground Truth but not Predicted):")
            for i, item in enumerate(comp_errors['false_negatives'][:10]):  # Show first 10
                print(f"      {i+1}. {item}")
            if len(comp_errors['false_negatives']) > 10:
                print(f"      ... and {len(comp_errors['false_negatives'])-10} more")

def print_component_summary_table(results: Dict):
    """Print a summary table comparing all components"""
    
    print(f"\n📊 COMPONENT PERFORMANCE SUMMARY:")
    print(f"{'Component':<20} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'GT Total':<8} | {'Pred Total':<10}")
    print("-" * 85)
    
    components_data = [
        ('Full Relations', results['overall_relations']),
        ('Sources', results['overall_sources']), 
        ('Targets', results['overall_targets']),
        ('Relation Types', results['overall_relation_types'])
    ]
    
    for comp_name, metrics in components_data:
        print(f"{comp_name:<20} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
              f"{metrics['f1']:<9.3f} | {metrics['total_gt']:<8} | {metrics['total_pred']:<10}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "LabelledRelations_Increase_Decrease - Feuille 1.csv"
PREDICTIONS_CSV = "relations_table_20250709_165543 - claude ai pipline approtch (1).csv"

if __name__ == "__main__":
    try:
        # Run comprehensive evaluation with individual components
        results = evaluate_individual_components(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print summary table
        print_component_summary_table(results)
        
        # Print detailed report
        print_component_evaluation_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "comprehensive_evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json
import numpy as np

def print_metrics_definitions():
    """Print clear definitions of precision, recall, and F1-score"""
    print("\n" + "="*80)
    print("📚 EVALUATION METRICS DEFINITIONS")
    print("="*80)
    
    print("\n🎯 CONFUSION MATRIX COMPONENTS:")
    print("   • True Positives (TP):  Items correctly predicted as positive")
    print("   • False Positives (FP): Items incorrectly predicted as positive")
    print("   • False Negatives (FN): Items incorrectly predicted as negative")
    print("   • True Negatives (TN):  Items correctly predicted as negative")
    
    print("\n📊 EVALUATION METRICS:")
    print("   • PRECISION = TP / (TP + FP)")
    print("     → Of all items we predicted as positive, how many were actually positive?")
    print("     → Measures how accurate our positive predictions are")
    print("     → Range: [0, 1], Higher is better")
    
    print("\n   • RECALL = TP / (TP + FN)")
    print("     → Of all actual positive items, how many did we correctly identify?")
    print("     → Measures how complete our positive predictions are")
    print("     → Range: [0, 1], Higher is better")
    
    print("\n   • F1-SCORE = 2 × (Precision × Recall) / (Precision + Recall)")
    print("     → Harmonic mean of precision and recall")
    print("     → Balances precision and recall into a single metric")
    print("     → Range: [0, 1], Higher is better")
    
    print("\n💡 INTERPRETATION:")
    print("   • High Precision + Low Recall:  Conservative predictions (few false alarms, but miss many)")
    print("   • Low Precision + High Recall:  Liberal predictions (catch most, but many false alarms)")
    print("   • High Precision + High Recall: Ideal performance")
    print("   • Low Precision + Low Recall:   Poor performance")

class ConfusionMatrix:
    """Class to handle confusion matrix creation and export"""
    
    def __init__(self, component_name: str):
        self.component_name = component_name
        self.tp = 0
        self.fp = 0
        self.fn = 0
        self.tn = 0  # Note: TN is not typically used in relation extraction
        self.true_positives_list = []
        self.false_positives_list = []
        self.false_negatives_list = []
    
    def update(self, tp: int, fp: int, fn: int, tp_items: List = None, fp_items: List = None, fn_items: List = None):
        """Update confusion matrix counts"""
        self.tp += tp
        self.fp += fp
        self.fn += fn
        
        if tp_items:
            self.true_positives_list.extend(tp_items)
        if fp_items:
            self.false_positives_list.extend(fp_items)
        if fn_items:
            self.false_negatives_list.extend(fn_items)
    
    def get_metrics(self) -> Dict:
        """Calculate and return metrics"""
        precision = self.tp / (self.tp + self.fp) if (self.tp + self.fp) > 0 else 0.0
        recall = self.tp / (self.tp + self.fn) if (self.tp + self.fn) > 0 else 0.0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
        
        return {
            'component': self.component_name,
            'tp': self.tp,
            'fp': self.fp,
            'fn': self.fn,
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'total_predicted': self.tp + self.fp,
            'total_actual': self.tp + self.fn
        }
    
    def print_confusion_matrix(self):
        """Print confusion matrix in tabular format"""
        print(f"\n🎯 CONFUSION MATRIX - {self.component_name.upper()}")
        print("="*50)
        print("                    PREDICTED")
        print("                 Pos      Neg")
        print("ACTUAL    Pos   {:4d}     {:4d}    (Total: {})".format(self.tp, self.fn, self.tp + self.fn))
        print("          Neg   {:4d}      N/A    (Total: {})".format(self.fp, self.fp))
        print("              (Total: {})".format(self.tp + self.fp))
        
        metrics = self.get_metrics()
        print(f"\n📊 Metrics:")
        print(f"   Precision: {metrics['precision']:.3f} = {self.tp}/({self.tp}+{self.fp})")
        print(f"   Recall:    {metrics['recall']:.3f} = {self.tp}/({self.tp}+{self.fn})")
        print(f"   F1-Score:  {metrics['f1']:.3f}")
    
    def export_to_dict(self) -> Dict:
        """Export confusion matrix data to dictionary"""
        metrics = self.get_metrics()
        return {
            'confusion_matrix': {
                'component': self.component_name,
                'matrix': {
                    'true_positives': self.tp,
                    'false_positives': self.fp,
                    'false_negatives': self.fn
                },
                'metrics': {
                    'precision': metrics['precision'],
                    'recall': metrics['recall'],
                    'f1_score': metrics['f1']
                },
                'detailed_items': {
                    'true_positives': self.true_positives_list[:50],  # Limit for readability
                    'false_positives': self.false_positives_list[:50],
                    'false_negatives': self.false_negatives_list[:50]
                }
            }
        }

def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]
    
    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence (RELATIONS ONLY)"""
    
    df = pd.read_csv(csv_path)
    
    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")
    
    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])
    
    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/SOURCE')
    if not target_col:
        missing_cols.append('target/TARGET')
    if not relation_col:
        missing_cols.append('relationship/relation/RELATION')
    if not sentence_id_col:
        missing_cols.append('sentence_id/SENTENCE_ID/id')
    if not sentence_col:
        missing_cols.append('sentence/text/TEXT_SEGMENT')
    
    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")
    
    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")
    
    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {
        'relations': set(),
        'sources': set(),
        'targets': set(), 
        'relation_types': set()
    })
    
    for _, row in df.iterrows():
        sentence_id = row[sentence_id_col]
        sentence_text = row[sentence_col]
        source = str(row[source_col]).strip()
        target = str(row[target_col]).strip()
        relation = str(row[relation_col]).strip()

        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract full relations (source, relation, target)
        full_relation = (source, relation, target)
        sentence_data[sentence_id]['relations'].add(full_relation)
        
        # Extract individual components
        sentence_data[sentence_id]['sources'].add(source)
        sentence_data[sentence_id]['targets'].add(target)
        sentence_data[sentence_id]['relation_types'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics_with_items(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int, List, List, List]:
    """Calculate precision, recall, F1 and return items for confusion matrix"""
    
    tp_items = list(true_set & pred_set)  # True positives
    fp_items = list(pred_set - true_set)  # False positives
    fn_items = list(true_set - pred_set)  # False negatives
    
    tp = len(tp_items)
    fp = len(fp_items)
    fn = len(fn_items)
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn, tp_items, fp_items, fn_items

def analyze_component_errors(gt_components: Set, pred_components: Set, component_name: str) -> Dict:
    """Analyze component-level errors"""
    
    errors = {
        'false_positives': list(pred_components - gt_components),
        'false_negatives': list(gt_components - pred_components),
        'component_name': component_name
    }
    
    return errors

def evaluate_individual_components_with_confusion_matrix(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files with confusion matrices"""
    
    # Print metrics definitions first
    print_metrics_definitions()
    
    print("\n📂 Loading data...")
    gt_data = load_csv_relations_only(ground_truth_csv)
    pred_data = load_csv_relations_only(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize confusion matrices for each component
    confusion_matrices = {
        'relations': ConfusionMatrix('Full Relations'),
        'sources': ConfusionMatrix('Sources'),
        'targets': ConfusionMatrix('Targets'),
        'relation_types': ConfusionMatrix('Relation Types')
    }
    
    # Initialize results
    results = {
        'overall_relations': {},
        'overall_sources': {},
        'overall_targets': {},
        'overall_relation_types': {},
        'relation_metrics': {},
        'per_sentence': [],
        'confusion_matrices': {},
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': {},
            'source_errors': {},
            'target_errors': {},
            'relation_type_errors': {}
        }
    }
    
    # Collect all components
    all_gt_relations = set()
    all_pred_relations = set()
    all_gt_sources = set()
    all_pred_sources = set()
    all_gt_targets = set()
    all_pred_targets = set()
    all_gt_relation_types = set()
    all_pred_relation_types = set()
    
    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        pred_sentence = pred_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
        
        # Evaluation for this sentence - FULL RELATIONS
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        confusion_matrices['relations'].update(relation_tp, relation_fp, relation_fn, tp_items, fp_items, fn_items)
        
        # Evaluation for this sentence - SOURCES ONLY
        gt_sources = gt_sentence['sources']
        pred_sources = pred_sentence['sources']
        source_p, source_r, source_f1, source_tp, source_fp, source_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_sources, pred_sources)
        sentence_result.update({
            'source_precision': source_p,
            'source_recall': source_r,
            'source_f1': source_f1,
            'source_tp': source_tp,
            'source_fp': source_fp,
            'source_fn': source_fn
        })
        confusion_matrices['sources'].update(source_tp, source_fp, source_fn, tp_items, fp_items, fn_items)
        
        # Evaluation for this sentence - TARGETS ONLY
        gt_targets = gt_sentence['targets']
        pred_targets = pred_sentence['targets']
        target_p, target_r, target_f1, target_tp, target_fp, target_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_targets, pred_targets)
        sentence_result.update({
            'target_precision': target_p,
            'target_recall': target_r,
            'target_f1': target_f1,
            'target_tp': target_tp,
            'target_fp': target_fp,
            'target_fn': target_fn
        })
        confusion_matrices['targets'].update(target_tp, target_fp, target_fn, tp_items, fp_items, fn_items)
        
        # Evaluation for this sentence - RELATION TYPES ONLY
        gt_rel_types = gt_sentence['relation_types']
        pred_rel_types = pred_sentence['relation_types']
        rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_rel_types, pred_rel_types)
        sentence_result.update({
            'relation_type_precision': rel_type_p,
            'relation_type_recall': rel_type_r,
            'relation_type_f1': rel_type_f1,
            'relation_type_tp': rel_type_tp,
            'relation_type_fp': rel_type_fp,
            'relation_type_fn': rel_type_fn
        })
        confusion_matrices['relation_types'].update(rel_type_tp, rel_type_fp, rel_type_fn, tp_items, fp_items, fn_items)
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        all_gt_sources.update(gt_sources)
        all_pred_sources.update(pred_sources)
        all_gt_targets.update(gt_targets)
        all_pred_targets.update(pred_targets)
        all_gt_relation_types.update(gt_rel_types)
        all_pred_relation_types.update(pred_rel_types)
        
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics and store in results
    print("\n📊 Calculating overall metrics...")
    
    # Store overall metrics
    results['overall_relations'] = confusion_matrices['relations'].get_metrics()
    results['overall_sources'] = confusion_matrices['sources'].get_metrics()
    results['overall_targets'] = confusion_matrices['targets'].get_metrics()
    results['overall_relation_types'] = confusion_matrices['relation_types'].get_metrics()
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn, _, _, _ = calculate_metrics_with_items(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis for all components
    results['error_analysis']['relation_errors'] = analyze_component_errors(all_gt_relations, all_pred_relations, 'relations')
    results['error_analysis']['source_errors'] = analyze_component_errors(all_gt_sources, all_pred_sources, 'sources')
    results['error_analysis']['target_errors'] = analyze_component_errors(all_gt_targets, all_pred_targets, 'targets')
    results['error_analysis']['relation_type_errors'] = analyze_component_errors(all_gt_relation_types, all_pred_relation_types, 'relation_types')
    
    # Export confusion matrices
    for component_name, cm in confusion_matrices.items():
        results['confusion_matrices'][component_name] = cm.export_to_dict()
    
    # Print confusion matrices
    for cm in confusion_matrices.values():
        cm.print_confusion_matrix()
    
    return results

def print_component_evaluation_report(results: Dict):
    """Print comprehensive evaluation report with individual components"""
    
    print("\n" + "="*80)
    print("🎯 COMPREHENSIVE RELATION EXTRACTION EVALUATION REPORT")
    print("="*80)
    
    # Overall performance for all components
    components = [
        ('FULL RELATIONS', results['overall_relations']),
        ('SOURCES ONLY', results['overall_sources']),
        ('TARGETS ONLY', results['overall_targets']),
        ('RELATION TYPES ONLY', results['overall_relation_types'])
    ]
    
    for component_name, metrics in components:
        print(f"\n📈 {component_name} PERFORMANCE:")
        print(f"   Precision: {metrics['precision']:.3f}")
        print(f"   Recall:    {metrics['recall']:.3f}")
        print(f"   F1-Score:  {metrics['f1']:.3f}")
        print(f"   Total GT:   {metrics.get('total_actual', 'N/A')}")
        print(f"   Total Pred: {metrics.get('total_predicted', 'N/A')}")
        print(f"   True Positives:  {metrics['tp']}")
        print(f"   False Positives: {metrics['fp']}")
        print(f"   False Negatives: {metrics['fn']}")
    
    # Relation performance by type
    print(f"\n🔗 FULL RELATION PERFORMANCE BY TYPE:")
    print(f"{'Relation Type':<25} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 75)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by F1 score (descending)
    relation_items.sort(key=lambda x: x[1]['f1'], reverse=True)
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<25} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<25} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    
    error_components = [
        ('Full Relations', errors['relation_errors']),
        ('Sources', errors['source_errors']),
        ('Targets', errors['target_errors']),
        ('Relation Types', errors['relation_type_errors'])
    ]
    
    for comp_name, comp_errors in error_components:
        print(f"   {comp_name} - FP: {len(comp_errors['false_positives'])}, FN: {len(comp_errors['false_negatives'])}")

def print_component_summary_table(results: Dict):
    """Print a summary table comparing all components"""
    
    print(f"\n📊 COMPONENT PERFORMANCE SUMMARY:")
    print(f"{'Component':<20} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'GT Total':<8} | {'Pred Total':<10}")
    print("-" * 85)
    
    components_data = [
        ('Full Relations', results['overall_relations']),
        ('Sources', results['overall_sources']), 
        ('Targets', results['overall_targets']),
        ('Relation Types', results['overall_relation_types'])
    ]
    
    for comp_name, metrics in components_data:
        print(f"{comp_name:<20} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
              f"{metrics['f1']:<9.3f} | {metrics.get('total_actual', 'N/A'):<8} | {metrics.get('total_predicted', 'N/A'):<10}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

def export_confusion_matrices_to_csv(results: Dict, output_dir: str = "."):
    """Export confusion matrices to separate CSV files"""
    import os
    
    for component_name, cm_data in results['confusion_matrices'].items():
        cm_info = cm_data['confusion_matrix']
        
        # Create confusion matrix data
        matrix_data = {
            'Metric': ['True Positives', 'False Positives', 'False Negatives', 'Precision', 'Recall', 'F1-Score'],
            'Value': [
                cm_info['matrix']['true_positives'],
                cm_info['matrix']['false_positives'], 
                cm_info['matrix']['false_negatives'],
                f"{cm_info['metrics']['precision']:.3f}",
                f"{cm_info['metrics']['recall']:.3f}",
                f"{cm_info['metrics']['f1_score']:.3f}"
            ]
        }
        
        # Save to CSV
        df = pd.DataFrame(matrix_data)
        filename = f"confusion_matrix_{component_name.lower().replace(' ', '_')}.csv"
        filepath = os.path.join(output_dir, filename)
        df.to_csv(filepath, index=False)
        print(f"📊 Confusion matrix for {component_name} saved to: {filepath}")

# Set your file paths here
GROUND_TRUTH_CSV = "LabelledRelations_Increase_Decrease - Feuille 1.csv"
PREDICTIONS_CSV = "relations_table_20250709_165543 - claude ai pipline approtch (1).csv"

if __name__ == "__main__":
    try:
        # Run comprehensive evaluation with confusion matrices
        results = evaluate_individual_components_with_confusion_matrix(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print summary table
        print_component_summary_table(results)
        
        # Print detailed report
        print_component_evaluation_report(results)
        
        # Save results to JSON (includes confusion matrices)
        save_results_to_json(results, "comprehensive_evaluation_results_with_confusion_matrices.json")
        
        # Export confusion matrices to separate CSV files
        export_confusion_matrices_to_csv(results)
        
        print(f"\n✅ Evaluation complete! Check the output files for detailed confusion matrices.")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json


def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]
    
    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence (RELATIONS ONLY)"""
    
    df = pd.read_csv(csv_path)
    
    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")
    
    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])
    
    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/SOURCE')
    if not target_col:
        missing_cols.append('target/TARGET')
    if not relation_col:
        missing_cols.append('relationship/relation/RELATION')
    if not sentence_id_col:
        missing_cols.append('sentence_id/SENTENCE_ID/id')
    if not sentence_col:
        missing_cols.append('sentence/text/TEXT_SEGMENT')
    
    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")
    
    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")
    
    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {
        'relations': set(),
        'sources': set(),
        'targets': set(), 
        'relation_types': set()
    })
    
    for _, row in df.iterrows():
        sentence_id = row[sentence_id_col]
        sentence_text = row[sentence_col]
        source = str(row[source_col]).strip()
        target = str(row[target_col]).strip()
        relation = str(row[relation_col]).strip()

        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract full relations (source, relation, target)
        full_relation = (source, relation, target)
        sentence_data[sentence_id]['relations'].add(full_relation)
        
        # Extract individual components
        sentence_data[sentence_id]['sources'].add(source)
        sentence_data[sentence_id]['targets'].add(target)
        sentence_data[sentence_id]['relation_types'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_component_errors(gt_components: Set, pred_components: Set, component_name: str) -> Dict:
    """Analyze component-level errors"""
    
    errors = {
        'false_positives': list(pred_components - gt_components),
        'false_negatives': list(gt_components - pred_components),
        'component_name': component_name
    }
    
    return errors

def evaluate_individual_components(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files with individual component analysis"""
    
    print("📂 Loading data...")
    gt_data = load_csv_relations_only(ground_truth_csv)
    pred_data = load_csv_relations_only(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results
    results = {
        'overall_relations': {},
        'overall_sources': {},
        'overall_targets': {},
        'overall_relation_types': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': {},
            'source_errors': {},
            'target_errors': {},
            'relation_type_errors': {}
        }
    }
    
    # Collect all components
    all_gt_relations = set()
    all_pred_relations = set()
    all_gt_sources = set()
    all_pred_sources = set()
    all_gt_targets = set()
    all_pred_targets = set()
    all_gt_relation_types = set()
    all_pred_relation_types = set()
    
    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        pred_sentence = pred_data.get(sentence_id, {
            'relations': set(), 'sources': set(), 
            'targets': set(), 'relation_types': set()
        })
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
        
        # Evaluation for this sentence - FULL RELATIONS
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        # Evaluation for this sentence - SOURCES ONLY
        gt_sources = gt_sentence['sources']
        pred_sources = pred_sentence['sources']
        source_p, source_r, source_f1, source_tp, source_fp, source_fn = calculate_metrics(gt_sources, pred_sources)
        sentence_result.update({
            'source_precision': source_p,
            'source_recall': source_r,
            'source_f1': source_f1,
            'source_tp': source_tp,
            'source_fp': source_fp,
            'source_fn': source_fn
        })
        
        # Evaluation for this sentence - TARGETS ONLY
        gt_targets = gt_sentence['targets']
        pred_targets = pred_sentence['targets']
        target_p, target_r, target_f1, target_tp, target_fp, target_fn = calculate_metrics(gt_targets, pred_targets)
        sentence_result.update({
            'target_precision': target_p,
            'target_recall': target_r,
            'target_f1': target_f1,
            'target_tp': target_tp,
            'target_fp': target_fp,
            'target_fn': target_fn
        })
        
        # Evaluation for this sentence - RELATION TYPES ONLY
        gt_rel_types = gt_sentence['relation_types']
        pred_rel_types = pred_sentence['relation_types']
        rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn = calculate_metrics(gt_rel_types, pred_rel_types)
        sentence_result.update({
            'relation_type_precision': rel_type_p,
            'relation_type_recall': rel_type_r,
            'relation_type_f1': rel_type_f1,
            'relation_type_tp': rel_type_tp,
            'relation_type_fp': rel_type_fp,
            'relation_type_fn': rel_type_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        all_gt_sources.update(gt_sources)
        all_pred_sources.update(pred_sources)
        all_gt_targets.update(gt_targets)
        all_pred_targets.update(pred_targets)
        all_gt_relation_types.update(gt_rel_types)
        all_pred_relation_types.update(pred_rel_types)
        
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall metrics...")
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall_relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn,
        'total_gt': len(all_gt_relations),
        'total_pred': len(all_pred_relations)
    }
    
    # Overall source metrics
    source_p, source_r, source_f1, source_tp, source_fp, source_fn = calculate_metrics(all_gt_sources, all_pred_sources)
    results['overall_sources'] = {
        'precision': source_p, 'recall': source_r, 'f1': source_f1,
        'tp': source_tp, 'fp': source_fp, 'fn': source_fn,
        'total_gt': len(all_gt_sources),
        'total_pred': len(all_pred_sources)
    }
    
    # Overall target metrics
    target_p, target_r, target_f1, target_tp, target_fp, target_fn = calculate_metrics(all_gt_targets, all_pred_targets)
    results['overall_targets'] = {
        'precision': target_p, 'recall': target_r, 'f1': target_f1,
        'tp': target_tp, 'fp': target_fp, 'fn': target_fn,
        'total_gt': len(all_gt_targets),
        'total_pred': len(all_pred_targets)
    }
    
    # Overall relation type metrics
    rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn = calculate_metrics(all_gt_relation_types, all_pred_relation_types)
    results['overall_relation_types'] = {
        'precision': rel_type_p, 'recall': rel_type_r, 'f1': rel_type_f1,
        'tp': rel_type_tp, 'fp': rel_type_fp, 'fn': rel_type_fn,
        'total_gt': len(all_gt_relation_types),
        'total_pred': len(all_pred_relation_types)
    }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis for all components
    results['error_analysis']['relation_errors'] = analyze_component_errors(all_gt_relations, all_pred_relations, 'relations')
    results['error_analysis']['source_errors'] = analyze_component_errors(all_gt_sources, all_pred_sources, 'sources')
    results['error_analysis']['target_errors'] = analyze_component_errors(all_gt_targets, all_pred_targets, 'targets')
    results['error_analysis']['relation_type_errors'] = analyze_component_errors(all_gt_relation_types, all_pred_relation_types, 'relation_types')
    
    return results

def print_component_evaluation_report(results: Dict):
    """Print comprehensive evaluation report with individual components"""
    
    print("\n" + "="*80)
    print("🎯 COMPREHENSIVE RELATION EXTRACTION EVALUATION REPORT")
    print("="*80)
    
    # Overall performance for all components
    components = [
        ('FULL RELATIONS', results['overall_relations']),
        ('SOURCES ONLY', results['overall_sources']),
        ('TARGETS ONLY', results['overall_targets']),
        ('RELATION TYPES ONLY', results['overall_relation_types'])
    ]
    
    for component_name, metrics in components:
        print(f"\n📈 {component_name} PERFORMANCE:")
        print(f"   Precision: {metrics['precision']:.3f}")
        print(f"   Recall:    {metrics['recall']:.3f}")
        print(f"   F1-Score:  {metrics['f1']:.3f}")
        print(f"   Total GT:   {metrics['total_gt']}")
        print(f"   Total Pred: {metrics['total_pred']}")
        print(f"   True Positives:  {metrics['tp']}")
        print(f"   False Positives: {metrics['fp']}")
        print(f"   False Negatives: {metrics['fn']}")
    
    # Relation performance by type
    print(f"\n🔗 FULL RELATION PERFORMANCE BY TYPE:")
    print(f"{'Relation Type':<25} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 75)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by F1 score (descending)
    relation_items.sort(key=lambda x: x[1]['f1'], reverse=True)
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<25} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<25} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    
    error_components = [
        ('Full Relations', errors['relation_errors']),
        ('Sources', errors['source_errors']),
        ('Targets', errors['target_errors']),
        ('Relation Types', errors['relation_type_errors'])
    ]
    
    for comp_name, comp_errors in error_components:
        print(f"   {comp_name} - FP: {len(comp_errors['false_positives'])}, FN: {len(comp_errors['false_negatives'])}")
    
    # Show detailed errors for each component
    for comp_name, comp_errors in error_components:
        print(f"\n🔍 DETAILED {comp_name.upper()} ERRORS:")
        
        if comp_errors['false_positives']:
            print(f"\n   FALSE POSITIVES (Predicted but not in Ground Truth):")
            for i, item in enumerate(comp_errors['false_positives'][:10]):  # Show first 10
                print(f"      {i+1}. {item}")
            if len(comp_errors['false_positives']) > 10:
                print(f"      ... and {len(comp_errors['false_positives'])-10} more")
        
        if comp_errors['false_negatives']:
            print(f"\n   FALSE NEGATIVES (In Ground Truth but not Predicted):")
            for i, item in enumerate(comp_errors['false_negatives'][:10]):  # Show first 10
                print(f"      {i+1}. {item}")
            if len(comp_errors['false_negatives']) > 10:
                print(f"      ... and {len(comp_errors['false_negatives'])-10} more")

def print_component_summary_table(results: Dict):
    """Print a summary table comparing all components"""
    
    print(f"\n📊 COMPONENT PERFORMANCE SUMMARY:")
    print(f"{'Component':<20} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'GT Total':<8} | {'Pred Total':<10}")
    print("-" * 85)
    
    components_data = [
        ('Full Relations', results['overall_relations']),
        ('Sources', results['overall_sources']), 
        ('Targets', results['overall_targets']),
        ('Relation Types', results['overall_relation_types'])
    ]
    
    for comp_name, metrics in components_data:
        print(f"{comp_name:<20} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
              f"{metrics['f1']:<9.3f} | {metrics['total_gt']:<8} | {metrics['total_pred']:<10}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "LabelledRelations_Increase_Decrease - Feuille 1.csv"
PREDICTIONS_CSV = "relations_table_20250709_165543 - claude ai pipline approtch (1).csv"

if __name__ == "__main__":
    try:
        # Run comprehensive evaluation with individual components
        results = evaluate_individual_components(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print summary table
        print_component_summary_table(results)
        
        # Print detailed report
        print_component_evaluation_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "comprehensive_evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set, Any
import json
import numpy as np

def print_metrics_definitions():
    """Print clear definitions of precision, recall, and F1-score"""
    print("\n" + "="*80)
    print("📚 EVALUATION METRICS DEFINITIONS")
    print("="*80)

    print("\n🎯 CONFUSION MATRIX COMPONENTS:")
    print("   • True Positives (TP):  Items correctly predicted as positive")
    print("   • False Positives (FP): Items incorrectly predicted as positive")
    print("   • False Negatives (FN): Items incorrectly predicted as negative")
    print("   • True Negatives (TN):  Items correctly predicted as negative")

    print("\n📊 EVALUATION METRICS:")
    print("   • PRECISION = TP / (TP + FP)")
    print("     → Of all items we predicted as positive, how many were actually positive?")
    print("     → Measures how accurate our positive predictions are")
    print("     → Range: [0, 1], Higher is better")

    print("\n   • RECALL = TP / (TP + FN)")
    print("     → Of all actual positive items, how many did we correctly identify?")
    print("     → Measures how complete our positive predictions are")
    print("     → Range: [0, 1], Higher is better")

    print("\n   • F1-SCORE = 2 × (Precision × Recall) / (Precision + Recall)")
    print("     → Harmonic mean of precision and recall")
    print("     → Balances precision and recall into a single metric")
    print("     → Range: [0, 1], Higher is better")

    print("\n💡 INTERPRETATION:")
    print("   • High Precision + Low Recall:  Conservative predictions (few false alarms, but miss many)")
    print("   • Low Precision + High Recall:  Liberal predictions (catch most, but many false alarms)")
    print("   • High Precision + High Recall: Ideal performance")
    print("   • Low Precision + Low Recall:   Poor performance")

class ConfusionMatrix:
    """Class to handle confusion matrix creation and export"""

    def __init__(self, component_name: str):
        self.component_name = component_name
        self.tp = 0
        self.fp = 0
        self.fn = 0
        self.tn = 0  # Note: TN is not typically used in relation extraction
        # Store items as (item, sentence_id) tuples
        self.true_positives_list: List[Tuple[Any, str]] = []
        self.false_positives_list: List[Tuple[Any, str]] = []
        self.false_negatives_list: List[Tuple[Any, str]] = []

    def update(self, tp: int, fp: int, fn: int,
               tp_items_with_ids: List[Tuple[Any, str]] = None,
               fp_items_with_ids: List[Tuple[Any, str]] = None,
               fn_items_with_ids: List[Tuple[Any, str]] = None):
        """
        Update confusion matrix counts and store items along with their sentence IDs.
        """
        self.tp += tp
        self.fp += fp
        self.fn += fn

        if tp_items_with_ids:
            self.true_positives_list.extend(tp_items_with_ids)
        if fp_items_with_ids:
            self.false_positives_list.extend(fp_items_with_ids)
        if fn_items_with_ids:
            self.false_negatives_list.extend(fn_items_with_ids)

    def get_metrics(self) -> Dict:
        """Calculate and return metrics"""
        precision = self.tp / (self.tp + self.fp) if (self.tp + self.fp) > 0 else 0.0
        recall = self.tp / (self.tp + self.fn) if (self.tp + self.fn) > 0 else 0.0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

        return {
            'component': self.component_name,
            'tp': self.tp,
            'fp': self.fp,
            'fn': self.fn,
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'total_predicted': self.tp + self.fp,
            'total_actual': self.tp + self.fn
        }

    def print_confusion_matrix(self):
        """Print confusion matrix in tabular format"""
        print(f"\n🎯 CONFUSION MATRIX - {self.component_name.upper()}")
        print("="*50)
        print("                    PREDICTED")
        print("                 Pos      Neg")
        print("ACTUAL    Pos   {:4d}     {:4d}    (Total: {})".format(self.tp, self.fn, self.tp + self.fn))
        print("          Neg   {:4d}      N/A    (Total: {})".format(self.fp, self.fp))
        print("              (Total: {})".format(self.tp + self.fp))

        metrics = self.get_metrics()
        print(f"\n📊 Metrics:")
        print(f"   Precision: {metrics['precision']:.3f} = {self.tp}/({self.tp}+{self.fp})")
        print(f"   Recall:    {metrics['recall']:.3f} = {self.tp}/({self.tp}+{self.fn})")
        print(f"   F1-Score:  {metrics['f1']:.3f}")

    def export_to_dict(self) -> Dict:
        """Export confusion matrix data to dictionary"""
        metrics = self.get_metrics()
        return {
            'confusion_matrix': {
                'component': self.component_name,
                'matrix': {
                    'true_positives': self.tp,
                    'false_positives': self.fp,
                    'false_negatives': self.fn
                },
                'metrics': {
                    'precision': metrics['precision'],
                    'recall': metrics['recall'],
                    'f1_score': metrics['f1']
                },
                # Store (item, sentence_id) for detailed analysis
                'detailed_items': {
                    'true_positives': self.true_positives_list,
                    'false_positives': self.false_positives_list,
                    'false_negatives': self.false_negatives_list
                }
            }
        }
```

```python
def get_column_name(df, possible_names):
    """Find the correct column name from a list of possibilities"""
    # Strip spaces from column names for comparison
    df_columns_stripped = [(col.strip().lower(), col) for col in df.columns]

    for name in possible_names:
        name_lower = name.lower()
        for stripped_col_lower, original_col in df_columns_stripped:
            if stripped_col_lower == name_lower:
                return original_col
    return None

def load_csv_relations_only(csv_path: str) -> Tuple[Dict, Dict[str, str]]:
    """
    Load CSV data and organize by sentence (RELATIONS ONLY)
    Returns:
        Tuple[Dict, Dict[str, str]]: (sentence_data, sentence_id_to_text_map)
    """

    df = pd.read_csv(csv_path)

    # Print available columns for debugging
    print(f"📋 Available columns in {csv_path}:")
    for i, col in enumerate(df.columns):
        print(f"   {i}: '{col}' (stripped: '{col.strip()}')")

    # Try to find the correct column names
    source_col = get_column_name(df, ['source', 'SOURCE', 'Source', 'entity_1', 'ENTITY_1'])
    target_col = get_column_name(df, ['target', 'TARGET', 'Target', 'entity_2', 'ENTITY_2'])
    relation_col = get_column_name(df, ['relationship', 'RELATIONSHIP', 'Relationship', 'relation', 'RELATION', 'Relation', 'relation_type', 'RELATION_TYPE'])
    sentence_id_col = get_column_name(df, ['sentence_id', 'SENTENCE_ID', 'Sentence_ID', 'id', 'ID', 'doc_id', 'DOC_ID'])
    sentence_col = get_column_name(df, ['sentence', 'SENTENCE', 'Sentence', 'text', 'TEXT', 'Text', 'text_segment', 'TEXT_SEGMENT', 'Text_Segment'])

    # Check if we found all required columns
    missing_cols = []
    if not source_col:
        missing_cols.append('source/entity_1')
    if not target_col:
        missing_cols.append('target/entity_2')
    if not relation_col:
        missing_cols.append('relationship/relation_type')
    if not sentence_id_col:
        missing_cols.append('sentence_id/id/doc_id')
    if not sentence_col:
        missing_cols.append('sentence/text/text_segment')

    if missing_cols:
        print("Available columns after stripping:")
        for col in df.columns:
            print(f"  '{col.strip()}'")
        raise ValueError(f"❌ Could not find required columns: {missing_cols}")

    print(f"✅ Using columns:")
    print(f"   Source: '{source_col}'")
    print(f"   Target: '{target_col}'")
    print(f"   Relation: '{relation_col}'")
    print(f"   Sentence ID: '{sentence_id_col}'")
    print(f"   Sentence: '{sentence_col}'")

    # Group data by sentence - RELATIONS ONLY
    sentence_data = defaultdict(lambda: {
        'relations': set(),
        'sources': set(),
        'targets': set(),
        'relation_types': set()
    })
    sentence_id_to_text_map: Dict[str, str] = {}

    for _, row in df.iterrows():
        sentence_id = str(row[sentence_id_col]).strip() # Ensure string type
        sentence_text = str(row[sentence_col]).strip()
        source = str(row[source_col]).strip()
        target = str(row[target_col]).strip()
        relation = str(row[relation_col]).strip()

        sentence_data[sentence_id]['sentence'] = sentence_text
        sentence_id_to_text_map[sentence_id] = sentence_text

        # Extract full relations (source, relation, target)
        full_relation = (source, relation, target)
        sentence_data[sentence_id]['relations'].add(full_relation)

        # Extract individual components
        sentence_data[sentence_id]['sources'].add(source)
        sentence_data[sentence_id]['targets'].add(target)
        sentence_data[sentence_id]['relation_types'].add(relation)

    return dict(sentence_data), sentence_id_to_text_map
```

```python
def calculate_metrics_with_items(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int, List, List, List]:
    """Calculate precision, recall, F1 and return items for confusion matrix"""

    tp_items = list(true_set & pred_set)  # True positives
    fp_items = list(pred_set - true_set)  # False positives
    fn_items = list(true_set - pred_set)  # False negatives

    tp = len(tp_items)
    fp = len(fp_items)
    fn = len(fn_items)

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

    return precision, recall, f1, tp, fp, fn, tp_items, fp_items, fn_items

def analyze_component_errors(gt_components: Set, pred_components: Set, component_name: str) -> Dict:
    """Analyze component-level errors"""

    errors = {
        'false_positives': list(pred_components - gt_components),
        'false_negatives': list(gt_components - pred_components),
        'component_name': component_name
    }

    return errors

def evaluate_individual_components_with_confusion_matrix(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files with confusion matrices"""

    # Print metrics definitions first
    print_metrics_definitions()

    print("\n📂 Loading data...")
    gt_data, gt_sentences_map = load_csv_relations_only(ground_truth_csv)
    pred_data, pred_sentences_map = load_csv_relations_only(predictions_csv)

    # Merge sentence text maps from both ground truth and predictions
    # This ensures we have the sentence text for any sentence_id present in either dataset
    all_sentences_map = {**gt_sentences_map, **pred_sentences_map}


    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")

    # Initialize confusion matrices for each component
    confusion_matrices = {
        'relations': ConfusionMatrix('Full Relations'),
        'sources': ConfusionMatrix('Sources'),
        'targets': ConfusionMatrix('Targets'),
        'relation_types': ConfusionMatrix('Relation Types')
    }

    # Initialize results
    results = {
        'overall_relations': {},
        'overall_sources': {},
        'overall_targets': {},
        'overall_relation_types': {},
        'relation_metrics': {},
        'per_sentence': [],
        'confusion_matrices': {},
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'relation_errors': {},
            'source_errors': {},
            'target_errors': {},
            'relation_type_errors': {}
        },
        'all_sentences_map': all_sentences_map # Add sentence map to results
    }

    # Collect all components
    all_gt_relations = set()
    all_pred_relations = set()
    all_gt_sources = set()
    all_pred_sources = set()
    all_gt_targets = set()
    all_pred_targets = set()
    all_gt_relation_types = set()
    all_pred_relation_types = set()

    # Collect relations by type
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})

    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())

    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")

    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}

        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {
            'relations': set(), 'sources': set(),
            'targets': set(), 'relation_types': set()
        })
        pred_sentence = pred_data.get(sentence_id, {
            'relations': set(), 'sources': set(),
            'targets': set(), 'relation_types': set()
        })

        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)

        # Evaluation for this sentence - FULL RELATIONS
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_relations, pred_relations)

        # Append sentence_id to each item before updating confusion matrix
        tp_relations_with_ids = [(item, sentence_id) for item in tp_items]
        fp_relations_with_ids = [(item, sentence_id) for item in fp_items]
        fn_relations_with_ids = [(item, sentence_id) for item in fn_items]

        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        confusion_matrices['relations'].update(relation_tp, relation_fp, relation_fn,
                                               tp_items_with_ids=tp_relations_with_ids,
                                               fp_items_with_ids=fp_relations_with_ids,
                                               fn_items_with_ids=fn_relations_with_ids)

        # Evaluation for this sentence - SOURCES ONLY
        gt_sources = gt_sentence['sources']
        pred_sources = pred_sentence['sources']
        source_p, source_r, source_f1, source_tp, source_fp, source_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_sources, pred_sources)

        tp_sources_with_ids = [(item, sentence_id) for item in tp_items]
        fp_sources_with_ids = [(item, sentence_id) for item in fp_items]
        fn_sources_with_ids = [(item, sentence_id) for item in fn_items]

        sentence_result.update({
            'source_precision': source_p,
            'source_recall': source_r,
            'source_f1': source_f1,
            'source_tp': source_tp,
            'source_fp': source_fp,
            'source_fn': source_fn
        })
        confusion_matrices['sources'].update(source_tp, source_fp, source_fn,
                                             tp_items_with_ids=tp_sources_with_ids,
                                             fp_items_with_ids=fp_sources_with_ids,
                                             fn_items_with_ids=fn_sources_with_ids)

        # Evaluation for this sentence - TARGETS ONLY
        gt_targets = gt_sentence['targets']
        pred_targets = pred_sentence['targets']
        target_p, target_r, target_f1, target_tp, target_fp, target_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_targets, pred_targets)

        tp_targets_with_ids = [(item, sentence_id) for item in tp_items]
        fp_targets_with_ids = [(item, sentence_id) for item in fp_items]
        fn_targets_with_ids = [(item, sentence_id) for item in fn_items]

        sentence_result.update({
            'target_precision': target_p,
            'target_recall': target_r,
            'target_f1': target_f1,
            'target_tp': target_tp,
            'target_fp': target_fp,
            'target_fn': target_fn
        })
        confusion_matrices['targets'].update(target_tp, target_fp, target_fn,
                                             tp_items_with_ids=tp_targets_with_ids,
                                             fp_items_with_ids=fp_targets_with_ids,
                                             fn_items_with_ids=fn_targets_with_ids)

        # Evaluation for this sentence - RELATION TYPES ONLY
        gt_rel_types = gt_sentence['relation_types']
        pred_rel_types = pred_sentence['relation_types']
        rel_type_p, rel_type_r, rel_type_f1, rel_type_tp, rel_type_fp, rel_type_fn, tp_items, fp_items, fn_items = calculate_metrics_with_items(gt_rel_types, pred_rel_types)

        tp_rel_types_with_ids = [(item, sentence_id) for item in tp_items]
        fp_rel_types_with_ids = [(item, sentence_id) for item in fp_items]
        fn_rel_types_with_ids = [(item, sentence_id) for item in fn_items]

        sentence_result.update({
            'relation_type_precision': rel_type_p,
            'relation_type_recall': rel_type_r,
            'relation_type_f1': rel_type_f1,
            'relation_type_tp': rel_type_tp,
            'relation_type_fp': rel_type_fp,
            'relation_type_fn': rel_type_fn
        })
        confusion_matrices['relation_types'].update(rel_type_tp, rel_type_fp, rel_type_fn,
                                                    tp_items_with_ids=tp_rel_types_with_ids,
                                                    fp_items_with_ids=fp_rel_types_with_ids,
                                                    fn_items_with_ids=fn_rel_types_with_ids)

        results['per_sentence'].append(sentence_result)

        # Accumulate for overall metrics
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        all_gt_sources.update(gt_sources)
        all_pred_sources.update(pred_sources)
        all_gt_targets.update(gt_targets)
        all_pred_targets.update(pred_targets)
        all_gt_relation_types.update(gt_rel_types)
        all_pred_relation_types.update(pred_rel_types)

        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[1]  # relation type is at index 1
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[1]
            relation_by_type[relation_type]['pred'].add(relation)

    # Calculate overall metrics and store in results
    print("\n📊 Calculating overall metrics...")

    # Store overall metrics
    results['overall_relations'] = confusion_matrices['relations'].get_metrics()
    results['overall_sources'] = confusion_matrices['sources'].get_metrics()
    results['overall_targets'] = confusion_matrices['targets'].get_metrics()
    results['overall_relation_types'] = confusion_matrices['relation_types'].get_metrics()

    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn, _, _, _ = calculate_metrics_with_items(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }

    # Error analysis for all components
    results['error_analysis']['relation_errors'] = analyze_component_errors(all_gt_relations, all_pred_relations, 'relations')
    results['error_analysis']['source_errors'] = analyze_component_errors(all_gt_sources, all_pred_sources, 'sources')
    results['error_analysis']['target_errors'] = analyze_component_errors(all_gt_targets, all_pred_targets, 'targets')
    results['error_analysis']['relation_type_errors'] = analyze_component_errors(all_gt_relation_types, all_pred_relation_types, 'relation_types')

    # Export confusion matrices
    for component_name, cm in confusion_matrices.items():
        results['confusion_matrices'][component_name] = cm.export_to_dict()

    # Print confusion matrices
    for cm in confusion_matrices.values():
        cm.print_confusion_matrix()

    return results
```
