```python
"""
LLaMA 3 Implementation for LULC Event Extraction - Complete Jupyter Notebook Version with Dates and JSON Output
========================================================================================================

This notebook demonstrates how to use LLaMA 3 for extracting Land Use Land Cover (LULC) 
change events from text, including temporal information and structured JSON output.

Requirements:
- Python 3.8+
- PyTorch 2.0+
- transformers 4.30.0+
- GPU with at least 16GB VRAM (for 8B model)
"""

# Cell 1: Import libraries and configure logging
import json
import os
import re
import logging
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from tqdm.notebook import tqdm
from datetime import datetime

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.StreamHandler()
    ]
)

print("Libraries imported successfully!")
```

```python
# Cell 3: Setup model and tokenizer
def setup_model(model_id, device, use_quantization=False):
    """
    Load the LLaMA 3 model and tokenizer.
    
    Args:
        model_id: Hugging Face model ID
        device: Device to load the model on
        use_quantization: Whether to use 4-bit quantization
        
    Returns:
        model, tokenizer
    """
    logging.info(f"Loading LLaMA 3 tokenizer: {model_id}")
    
    # Load tokenizer
    tokenizer = AutoTokenizer.from_pretrained(model_id)
    
    # Add padding token if it doesn't exist
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    
    # Configure quantization if requested
    if use_quantization:
        logging.info("Using 4-bit quantization")
        quantization_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_compute_dtype=torch.float16,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_use_double_quant=True
        )
    else:
        quantization_config = None
    
    # Load model with appropriate configuration
    logging.info(f"Loading LLaMA 3 model: {model_id}")
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        device_map="auto" if device == "cuda" else None,
        quantization_config=quantization_config,
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
        trust_remote_code=True
    )
    
    logging.info(f"LLaMA 3 model loaded successfully on {device}")
    return model, tokenizer

# Execute this cell to load the model
try:
    model, tokenizer = setup_model(MODEL_ID, DEVICE, USE_QUANTIZATION)
    print("✅ Model and tokenizer loaded successfully!")
except Exception as e:
    print(f"❌ Error loading model: {e}")
    model, tokenizer = None, None
```

```python
# Cell 4: Date extraction functions
def extract_dates_from_entities(entities):
    """
    Extract date information from entities.
    
    Args:
        entities: List of entity dictionaries
        
    Returns:
        Dictionary with date information
    """
    dates = []
    date_ranges = []
    
    # Extract DATE entities
    date_entities = [ent for ent in entities if ent.get('label') == 'DATE']
    
    for entity in date_entities:
        date_text = entity.get('text', '')
        if date_text:
            dates.append(date_text)
    
    # Try to identify date ranges
    if len(dates) >= 2:
        # Sort dates and create range
        dates_sorted = sorted(dates)
        date_ranges.append(f"{dates_sorted[0]} to {dates_sorted[-1]}")
    
    return {
        'dates': dates,
        'date_range': date_ranges[0] if date_ranges else "",
        'start_date': dates[0] if dates else "",
        'count': len(dates),
        'end_date': dates[-1] if len(dates) > 1 else ""
    }

def format_entities_for_prompt_with_dates(entities):
    """Enhanced version that formats entities for prompt inclusion."""
    if not entities:
        return "No specific entities pre-identified."
    
    # Group entities by type
    entities_by_type = {}
    for entity in entities:
        entity_type = entity.get('label', 'UNKNOWN')
        if entity_type not in entities_by_type:
            entities_by_type[entity_type] = []
        entities_by_type[entity_type].append(entity.get('text', 'N/A'))
    
    # Format grouped entities
    formatted_lines = []
    for entity_type, entity_texts in entities_by_type.items():
        unique_texts = list(set(entity_texts))  # Remove duplicates
        entity_list = ", ".join([f'"{text}"' for text in unique_texts])
        formatted_lines.append(f"- {entity_type}: {entity_list}")
    
    return "\n".join(formatted_lines)

print("✅ Date extraction functions defined!")
```

```python
# Cell 6: Enhanced text generation and parsing functions for comprehensive LULC relationships
import re
import json
import torch
from typing import Dict, List, Any, Optional, Tuple

def generate_lulc_relationships_with_llama3(model, tokenizer, prompt, max_new_tokens=512):
    """
    Generate comprehensive LULC relationship text using LLaMA 3 model.
    
    Args:
        model: LLaMA 3 model
        tokenizer: LLaMA 3 tokenizer
        prompt: Input prompt for relationship extraction
        max_new_tokens: Maximum number of tokens to generate (increased for complex output)
        
    Returns:
        Generated text containing comprehensive LULC relationships
    """
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=4000).to(model.device)
    
    # Generate with parameters optimized for structured relationship extraction
    with torch.no_grad():
        outputs = model.generate(
            inputs.input_ids,
            max_new_tokens=max_new_tokens,
            temperature=0.05,  # Very low temperature for consistent structured output
            top_p=0.95,
            do_sample=True,
            num_return_sequences=1,
            pad_token_id=tokenizer.eos_token_id,
            repetition_penalty=1.1,  # Prevent repetitive output
            length_penalty=1.0
        )
    
    # Decode and extract only the newly generated text
    full_output = tokenizer.decode(outputs[0], skip_special_tokens=True)
    generated_text = full_output[len(tokenizer.decode(inputs.input_ids[0], skip_special_tokens=True)):]
    
    # Clean up the output
    generated_text = generated_text.strip()
    
    # If the output contains closing tags, extract only the content before them
    end_markers = ["</output>", "</o>", "<end>", "---END---"]
    for marker in end_markers:
        if marker in generated_text:
            generated_text = generated_text.split(marker)[0].strip()
    
    return generated_text

def parse_comprehensive_lulc_output(output_text: str) -> Dict[str, Any]:
    """
    Enhanced parsing for comprehensive LULC relationship extraction.
    
    Args:
        output_text: Raw output from LLaMA 3 containing structured relationships
        
    Returns:
        Dictionary with comprehensive parsed LULC relationship fields
    """
    # Check for NO_RELATIONSHIP marker
    if "NO_RELATIONSHIP" in output_text or "NO_EVENT" in output_text:
        return {
            "relationship_found": False,
            "transformation": {},
            "spatial_context": {},
            "temporal_context": {},
            "causal_mechanisms": {},
            "transformation_rules": {},
            "effects_and_impacts": {},
            "relationships": {},
            "raw_output": output_text
        }
    
    # Parse each major section
    transformation = parse_transformation_section(output_text)
    spatial_context = parse_spatial_context_section(output_text)
    temporal_context = parse_temporal_context_section(output_text)
    causal_mechanisms = parse_causal_mechanisms_section(output_text)
    transformation_rules = parse_transformation_rules_section(output_text)
    effects_and_impacts = parse_effects_and_impacts_section(output_text)
    relationships = parse_relationships_section(output_text)
    
    # Determine if a relationship was found
    relationship_found = bool(
        transformation.get("from_lulc") or 
        transformation.get("to_lulc") or 
        transformation.get("change_type") or
        spatial_context.get("location") or
        causal_mechanisms.get("primary_drivers")
    )
    
    return {
        "relationship_found": relationship_found,
        "transformation": transformation,
        "spatial_context": spatial_context,
        "temporal_context": temporal_context,
        "causal_mechanisms": causal_mechanisms,
        "transformation_rules": transformation_rules,
        "effects_and_impacts": effects_and_impacts,
        "relationships": relationships,
        "raw_output": output_text
    }

def parse_transformation_section(text: str) -> Dict[str, str]:
    """Parse the TRANSFORMATION section."""
    section = extract_section(text, "TRANSFORMATION:")
    
    return {
        "from_lulc": extract_field_value(section, "FROM:"),
        "to_lulc": extract_field_value(section, "TO:"),
        "change_type": extract_field_value(section, "CHANGE_TYPE:"),
        "magnitude": extract_field_value(section, "MAGNITUDE:")
    }

def parse_spatial_context_section(text: str) -> Dict[str, str]:
    """Parse the SPATIAL_CONTEXT section."""
    section = extract_section(text, "SPATIAL_CONTEXT:")
    
    return {
        "location": extract_field_value(section, "LOCATION:"),
        "spatial_scale": extract_field_value(section, "SPATIAL_SCALE:"),
        "spatial_pattern": extract_field_value(section, "SPATIAL_PATTERN:")
    }

def parse_temporal_context_section(text: str) -> Dict[str, str]:
    """Parse the TEMPORAL_CONTEXT section."""
    section = extract_section(text, "TEMPORAL_CONTEXT:")
    
    return {
        "start_date": extract_field_value(section, "START_DATE:"),
        "end_date": extract_field_value(section, "END_DATE:"),
        "duration": extract_field_value(section, "DURATION:"),
        "change_rate": extract_field_value(section, "CHANGE_RATE:"),
        "temporal_pattern": extract_field_value(section, "TEMPORAL_PATTERN:")
    }

def parse_causal_mechanisms_section(text: str) -> Dict[str, str]:
    """Parse the CAUSAL_MECHANISMS section."""
    section = extract_section(text, "CAUSAL_MECHANISMS:")
    
    return {
        "primary_drivers": extract_field_value(section, "PRIMARY_DRIVERS:"),
        "secondary_drivers": extract_field_value(section, "SECONDARY_DRIVERS:"),
        "human_factors": extract_field_value(section, "HUMAN_FACTORS:"),
        "natural_factors": extract_field_value(section, "NATURAL_FACTORS:"),
        "policy_factors": extract_field_value(section, "POLICY_FACTORS:")
    }

def parse_transformation_rules_section(text: str) -> Dict[str, str]:
    """Parse the TRANSFORMATION_RULES section."""
    section = extract_section(text, "TRANSFORMATION_RULES:")
    
    return {
        "conditions": extract_field_value(section, "CONDITIONS:"),
        "prerequisites": extract_field_value(section, "PREREQUISITES:"),
        "constraints": extract_field_value(section, "CONSTRAINTS:"),
        "thresholds": extract_field_value(section, "THRESHOLDS:")
    }

def parse_effects_and_impacts_section(text: str) -> Dict[str, str]:
    """Parse the EFFECTS_AND_IMPACTS section."""
    section = extract_section(text, "EFFECTS_AND_IMPACTS:")
    
    return {
        "environmental_effects": extract_field_value(section, "ENVIRONMENTAL_EFFECTS:"),
        "socioeconomic_effects": extract_field_value(section, "SOCIOECONOMIC_EFFECTS:"),
        "ecosystem_services": extract_field_value(section, "ECOSYSTEM_SERVICES:"),
        "cascading_effects": extract_field_value(section, "CASCADING_EFFECTS:")
    }

def parse_relationships_section(text: str) -> Dict[str, str]:
    """Parse the RELATIONSHIPS section."""
    section = extract_section(text, "RELATIONSHIPS:")
    
    return {
        "causal_chain": extract_field_value(section, "CAUSAL_CHAIN:"),
        "feedback_loops": extract_field_value(section, "FEEDBACK_LOOPS:"),
        "spatial_dependencies": extract_field_value(section, "SPATIAL_DEPENDENCIES:"),
        "temporal_dependencies": extract_field_value(section, "TEMPORAL_DEPENDENCIES:")
    }

def extract_section(text: str, section_header: str) -> str:
    """Extract a specific section from the text."""
    # Find the section start
    section_start = text.find(section_header)
    if section_start == -1:
        return ""
    
    # Find the next major section or end of text
    next_sections = [
        "TRANSFORMATION:", "SPATIAL_CONTEXT:", "TEMPORAL_CONTEXT:",
        "CAUSAL_MECHANISMS:", "TRANSFORMATION_RULES:", "EFFECTS_AND_IMPACTS:",
        "RELATIONSHIPS:"
    ]
    
    section_end = len(text)
    for next_section in next_sections:
        if next_section != section_header:
            next_pos = text.find(next_section, section_start + len(section_header))
            if next_pos != -1 and next_pos < section_end:
                section_end = next_pos
    
    return text[section_start:section_end]

def extract_field_value(section: str, field_name: str) -> str:
    """Extract a field value from a section."""
    field_match = re.search(rf'{re.escape(field_name)}\s*(.*?)(?=\n[A-Z_]+:|$)', section, re.DOTALL)
    if field_match:
        value = field_match.group(1).strip()
        # Clean up common artifacts
        value = re.sub(r'^[-\s]*', '', value)  # Remove leading dashes and spaces
        value = re.sub(r'\s+', ' ', value)     # Normalize whitespace
        return value
    return ""

def process_magnitude_comprehensive(magnitude: str) -> Dict[str, str]:
    """
    Process magnitude into different components with enhanced parsing.
    
    Args:
        magnitude: Raw magnitude string
        
    Returns:
        Dictionary with different magnitude types
    """
    if not magnitude:
        return {
            "magnitude_percent": "",
            "magnitude_area": "",
            "magnitude_rate": "",
            "magnitude_absolute": ""
        }
    
    magnitude_percent = ""
    magnitude_area = ""
    magnitude_rate = ""
    magnitude_absolute = ""
    
    # Check for percentage
    percent_match = re.search(r'\d+\.?\d*\s*%', magnitude)
    if percent_match:
        magnitude_percent = percent_match.group()
    
    # Check for area units
    area_patterns = [
        r'\d+\.?\d*\s*(?:km²|km2|square\s*km|sq\s*km)',
        r'\d+\.?\d*\s*(?:ha|hectare)',
        r'\d+\.?\d*\s*(?:acre|acres)',
        r'\d+\.?\d*\s*(?:m²|m2|square\s*meter)'
    ]
    
    for pattern in area_patterns:
        area_match = re.search(pattern, magnitude, re.IGNORECASE)
        if area_match:
            magnitude_area = area_match.group()
            break
    
    # Check for rate indicators
    rate_patterns = [
        r'\d+\.?\d*\s*(?:%\s*per\s*year|%/year)',
        r'\d+\.?\d*\s*(?:km²\s*per\s*year|km²/year)',
        r'\d+\.?\d*\s*(?:ha\s*per\s*year|ha/year)'
    ]
    
    for pattern in rate_patterns:
        rate_match = re.search(pattern, magnitude, re.IGNORECASE)
        if rate_match:
            magnitude_rate = rate_match.group()
            break
    
    # Extract absolute numbers
    absolute_match = re.search(r'\d+\.?\d*', magnitude)
    if absolute_match:
        magnitude_absolute = absolute_match.group()
    
    return {
        "magnitude_percent": magnitude_percent,
        "magnitude_area": magnitude_area,
        "magnitude_rate": magnitude_rate,
        "magnitude_absolute": magnitude_absolute
    }

def extract_locations_hierarchy(location_text: str) -> Dict[str, List[str]]:
    """
    Extract location information and organize by geographic hierarchy.
    
    Args:
        location_text: Raw location text
        
    Returns:
        Dictionary with locations organized by hierarchy level
    """
    if not location_text:
        return {
            "countries": [],
            "regions": [],
            "cities": [],
            "landmarks": [],
            "coordinates": []
        }
    
    # Split multiple locations
    locations = re.split(r'[,;]|\sand\s', location_text)
    locations = [loc.strip() for loc in locations if loc.strip()]
    
    countries = []
    regions = []
    cities = []
    landmarks = []
    coordinates = []
    
    for location in locations:
        # Check for coordinates
        if re.search(r'\d+\.?\d*°?\s*[NS]\s*,?\s*\d+\.?\d*°?\s*[EW]', location):
            coordinates.append(location)
        # Check for known country patterns (this could be expanded with gazetteer)
        elif any(keyword in location.lower() for keyword in ['country', 'nation', 'republic']):
            countries.append(location)
        # Check for region patterns
        elif any(keyword in location.lower() for keyword in ['region', 'province', 'state', 'county']):
            regions.append(location)
        # Check for city patterns
        elif any(keyword in location.lower() for keyword in ['city', 'town', 'metropolitan', 'urban']):
            cities.append(location)
        # Check for landmark patterns
        elif any(keyword in location.lower() for keyword in ['forest', 'park', 'reserve', 'basin', 'valley']):
            landmarks.append(location)
        else:
            # Default categorization based on common geographic names
            cities.append(location)  # Default to cities for unclassified locations
    
    return {
        "countries": countries,
        "regions": regions, 
        "cities": cities,
        "landmarks": landmarks,
        "coordinates": coordinates
    }

def create_lulc_knowledge_graph_entry(parsed_data: Dict[str, Any]) -> Dict[str, Any]:
    """
    Create a knowledge graph entry from parsed LULC relationship data.
    
    Args:
        parsed_data: Comprehensive parsed LULC relationship data
        
    Returns:
        Knowledge graph entry format
    """
    if not parsed_data["relationship_found"]:
        return {"nodes": [], "edges": [], "relationship_type": "none"}
    
    nodes = []
    edges = []
    
    # Create nodes for LULC types
    transformation = parsed_data["transformation"]
    if transformation.get("from_lulc"):
        nodes.append({
            "id": f"lulc_{transformation['from_lulc'].replace(' ', '_')}",
            "type": "LULC",
            "label": transformation["from_lulc"],
            "role": "source"
        })
    
    if transformation.get("to_lulc"):
        nodes.append({
            "id": f"lulc_{transformation['to_lulc'].replace(' ', '_')}",
            "type": "LULC", 
            "label": transformation["to_lulc"],
            "role": "target"
        })
    
    # Create location nodes
    spatial = parsed_data["spatial_context"]
    if spatial.get("location"):
        nodes.append({
            "id": f"location_{spatial['location'].replace(' ', '_')}",
            "type": "LOCATION",
            "label": spatial["location"],
            "scale": spatial.get("spatial_scale", "")
        })
    
    # Create driver nodes
    causal = parsed_data["causal_mechanisms"]
    drivers = []
    if causal.get("primary_drivers"):
        drivers.extend(causal["primary_drivers"].split(", "))
    if causal.get("secondary_drivers"):
        drivers.extend(causal["secondary_drivers"].split(", "))
    
    for driver in drivers:
        if driver.strip():
            nodes.append({
                "id": f"driver_{driver.replace(' ', '_')}",
                "type": "DRIVER",
                "label": driver.strip()
            })
    
    # Create transformation edge
    if transformation.get("from_lulc") and transformation.get("to_lulc"):
        edges.append({
            "source": f"lulc_{transformation['from_lulc'].replace(' ', '_')}",
            "target": f"lulc_{transformation['to_lulc'].replace(' ', '_')}",
            "type": "TRANSFORMS_TO",
            "properties": {
                "change_type": transformation.get("change_type", ""),
                "magnitude": transformation.get("magnitude", ""),
                "temporal_context": parsed_data["temporal_context"]
            }
        })
    
    return {
        "nodes": nodes,
        "edges": edges,
        "relationship_type": transformation.get("change_type", "transformation"),
        "metadata": {
            "spatial_context": parsed_data["spatial_context"],
            "temporal_context": parsed_data["temporal_context"],
            "effects": parsed_data["effects_and_impacts"]
        }
    }

print("✅ Enhanced LULC relationship generation and parsing functions defined!")
print("📊 New capabilities:")
print("- Comprehensive relationship structure parsing")
print("- Enhanced spatial and temporal context extraction")
print("- Causal mechanism and transformation rules parsing")
print("- Effects and impacts analysis")
print("- Complex relationship mapping")
print("- Knowledge graph entry creation")
print("- Hierarchical location processing")
print("- Multi-dimensional magnitude processing")
```

```python
import json
import uuid
from datetime import datetime
from typing import Dict, List, Any, Optional

def convert_to_labelstudio_format(
    sentence_text: str,
    entities: List[Dict],
    parsed_relationships: Dict[str, Any],
    task_id: Optional[int] = None,
    include_predictions: bool = True,
    include_annotations: bool = True
) -> Dict[str, Any]:
    """
    Convert LULC relationship extraction results to Label Studio JSON format.
    
    Args:
        sentence_text: Original input sentence
        entities: List of extracted entities
        parsed_relationships: Comprehensive parsed LULC relationship data
        task_id: Unique task identifier
        include_predictions: Whether to include model predictions
        include_annotations: Whether to include ground truth annotations
        
    Returns:
        Label Studio formatted JSON structure
    """
    if task_id is None:
        task_id = hash(sentence_text) % 1000000
    
    # Base task structure
    labelstudio_task = {
        "id": task_id,
        "data": {
            "text": sentence_text,
            "meta": {
                "source": "lulc_relationship_extractor",
                "timestamp": datetime.now().isoformat(),
                "entity_count": len(entities),
                "relationship_found": parsed_relationships.get("relationship_found", False)
            }
        },
        "annotations": [],
        "predictions": [],
        "project": "lulc_relationship_extraction"
    }
    
    # Add annotations (ground truth or manual annotations)
    if include_annotations:
        annotation = create_annotation_structure(entities, parsed_relationships, sentence_text)
        labelstudio_task["annotations"].append(annotation)
    
    # Add predictions (model outputs)
    if include_predictions:
        prediction = create_prediction_structure(entities, parsed_relationships, sentence_text)
        labelstudio_task["predictions"].append(prediction)
    
    return labelstudio_task

def create_annotation_structure(
    entities: List[Dict],
    parsed_relationships: Dict[str, Any],
    sentence_text: str
) -> Dict[str, Any]:
    """Create Label Studio annotation structure."""
    
    annotation_id = str(uuid.uuid4())
    
    # Base annotation structure
    annotation = {
        "id": annotation_id,
        "created_username": "lulc_extractor",
        "created_ago": "0 minutes",
        "result": [],
        "was_cancelled": False,
        "ground_truth": False,
        "created_at": datetime.now().isoformat(),
        "updated_at": datetime.now().isoformat(),
        "lead_time": 0.0,
        "prediction": {},
        "result_count": 0,
        "task": None
    }
    
    # Add entity annotations (NER-style)
    entity_results = create_entity_annotations(entities, sentence_text)
    annotation["result"].extend(entity_results)
    
    # Add relationship annotations
    relationship_results = create_relationship_annotations(parsed_relationships)
    annotation["result"].extend(relationship_results)
    
    # Add classification results
    classification_results = create_classification_annotations(parsed_relationships)
    annotation["result"].extend(classification_results)
    
    annotation["result_count"] = len(annotation["result"])
    
    return annotation

def create_prediction_structure(
    entities: List[Dict],
    parsed_relationships: Dict[str, Any],
    sentence_text: str
) -> Dict[str, Any]:
    """Create Label Studio prediction structure."""
    
    prediction_id = str(uuid.uuid4())
    
    # Base prediction structure
    prediction = {
        "id": prediction_id,
        "created_ago": "0 minutes",
        "result": [],
        "score": calculate_confidence_score(parsed_relationships),
        "cluster": None,
        "neighbors": [],
        "mislabeling": 0.0,
        "created_at": datetime.now().isoformat(),
        "updated_at": datetime.now().isoformat(),
        "model_version": "lulc_llama3_v1.0"
    }
    
    # Add entity predictions
    entity_results = create_entity_annotations(entities, sentence_text, is_prediction=True)
    prediction["result"].extend(entity_results)
    
    # Add relationship predictions
    relationship_results = create_relationship_annotations(parsed_relationships, is_prediction=True)
    prediction["result"].extend(relationship_results)
    
    # Add classification predictions
    classification_results = create_classification_annotations(parsed_relationships, is_prediction=True)
    prediction["result"].extend(classification_results)
    
    return prediction

def create_entity_annotations(
    entities: List[Dict],
    sentence_text: str,
    is_prediction: bool = False
) -> List[Dict[str, Any]]:
    """Create entity annotations in Label Studio format."""
    
    results = []
    
    for i, entity in enumerate(entities):
        entity_text = entity.get('text', '')
        entity_label = entity.get('label', 'UNKNOWN')
        start_pos = entity.get('start', 0)
        end_pos = entity.get('end', len(entity_text))
        
        # Calculate character positions
        if start_pos is None or end_pos is None:
            # Try to find the entity in the text
            text_start = sentence_text.find(entity_text)
            if text_start != -1:
                start_pos = text_start
                end_pos = text_start + len(entity_text)
            else:
                start_pos = 0
                end_pos = len(entity_text)
        
        # Convert to Label Studio format (percentage-based positions)
        text_length = len(sentence_text)
        start_percent = (start_pos / text_length * 100) if text_length > 0 else 0
        end_percent = (end_pos / text_length * 100) if text_length > 0 else 0
        
        entity_result = {
            "id": f"entity_{i}_{uuid.uuid4().hex[:8]}",
            "type": "labels",
            "value": {
                "start": start_percent,
                "end": end_percent,
                "text": entity_text,
                "labels": [entity_label]
            },
            "to_name": "text",
            "from_name": "label",
            "original_length": len(entity_text),
            "origin": "prediction" if is_prediction else "manual"
        }
        
        if is_prediction:
            entity_result["score"] = calculate_entity_confidence(entity)
        
        results.append(entity_result)
    
    return results

def create_relationship_annotations(
    parsed_relationships: Dict[str, Any],
    is_prediction: bool = False
) -> List[Dict[str, Any]]:
    """Create relationship annotations in Label Studio format."""
    
    if not parsed_relationships.get("relationship_found", False):
        return []
    
    results = []
    
    # Create transformation relationship
    transformation = parsed_relationships.get("transformation", {})
    if transformation:
        relation_result = {
            "id": f"relation_{uuid.uuid4().hex[:8]}",
            "type": "relation",
            "value": {
                "from_id": "lulc_from",
                "to_id": "lulc_to",
                "type": "TRANSFORMS_TO",
                "direction": "right",
                "labels": [transformation.get("change_type", "transformation")]
            },
            "to_name": "label",
            "from_name": "relation",
            "origin": "prediction" if is_prediction else "manual"
        }
        
        if is_prediction:
            relation_result["score"] = 0.85
        
        results.append(relation_result)
    
    # Create causal relationships
    causal = parsed_relationships.get("causal_mechanisms", {})
    if causal.get("primary_drivers"):
        causal_result = {
            "id": f"causal_{uuid.uuid4().hex[:8]}",
            "type": "relation",
            "value": {
                "from_id": "driver",
                "to_id": "transformation",
                "type": "CAUSES",
                "direction": "right",
                "labels": ["causation"]
            },
            "to_name": "label",
            "from_name": "relation",
            "origin": "prediction" if is_prediction else "manual"
        }
        
        if is_prediction:
            causal_result["score"] = 0.80
        
        results.append(causal_result)
    
    return results

def create_classification_annotations(
    parsed_relationships: Dict[str, Any],
    is_prediction: bool = False
) -> List[Dict[str, Any]]:
    """Create classification annotations in Label Studio format."""
    
    results = []
    
    # Document-level classification
    doc_classification = {
        "id": f"classification_{uuid.uuid4().hex[:8]}",
        "type": "choices",
        "value": {
            "choices": []
        },
        "to_name": "text",
        "from_name": "document_type",
        "origin": "prediction" if is_prediction else "manual"
    }
    
    # Add relationship type classification
    if parsed_relationships.get("relationship_found", False):
        transformation = parsed_relationships.get("transformation", {})
        change_type = transformation.get("change_type", "")
        
        if change_type:
            doc_classification["value"]["choices"].append(f"LULC_{change_type.upper()}")
        else:
            doc_classification["value"]["choices"].append("LULC_TRANSFORMATION")
    else:
        doc_classification["value"]["choices"].append("NO_LULC_RELATIONSHIP")
    
    if is_prediction:
        doc_classification["score"] = 0.90 if parsed_relationships.get("relationship_found") else 0.95
    
    results.append(doc_classification)
    
    # Add spatial scale classification
    spatial = parsed_relationships.get("spatial_context", {})
    if spatial.get("spatial_scale"):
        spatial_classification = {
            "id": f"spatial_{uuid.uuid4().hex[:8]}",
            "type": "choices",
            "value": {
                "choices": [f"SCALE_{spatial['spatial_scale'].upper()}"]
            },
            "to_name": "text",
            "from_name": "spatial_scale",
            "origin": "prediction" if is_prediction else "manual"
        }
        
        if is_prediction:
            spatial_classification["score"] = 0.85
        
        results.append(spatial_classification)
    
    return results

def calculate_confidence_score(parsed_relationships: Dict[str, Any]) -> float:
    """Calculate overall confidence score for the prediction."""
    
    if not parsed_relationships.get("relationship_found", False):
        return 0.95  # High confidence in NO_RELATIONSHIP
    
    # Base score
    score = 0.5
    
    # Add points for different components
    transformation = parsed_relationships.get("transformation", {})
    if transformation.get("from_lulc"):
        score += 0.1
    if transformation.get("to_lulc"):
        score += 0.1
    if transformation.get("change_type"):
        score += 0.1
    
    spatial = parsed_relationships.get("spatial_context", {})
    if spatial.get("location"):
        score += 0.05
    
    temporal = parsed_relationships.get("temporal_context", {})
    if temporal.get("start_date") or temporal.get("end_date"):
        score += 0.05
    
    causal = parsed_relationships.get("causal_mechanisms", {})
    if causal.get("primary_drivers"):
        score += 0.1
    
    return min(score, 0.95)  # Cap at 95%

def calculate_entity_confidence(entity: Dict[str, Any]) -> float:
    """Calculate confidence score for individual entity."""
    
    # Base confidence based on entity type
    label = entity.get('label', '')
    base_scores = {
        'LULC': 0.9,
        'LOCATION': 0.85,
        'DATE': 0.95,
        'PERCENT': 0.9,
        'CHANGE': 0.8,
        'GPE': 0.85,
        'LOC': 0.85
    }
    
    return base_scores.get(label, 0.7)

def save_labelstudio_batch(
    results: List[Dict[str, Any]],
    output_file: str = "lulc_relationships_labelstudio.json"
) -> None:
    """
    Save multiple Label Studio tasks to a JSON file.
    
    Args:
        results: List of Label Studio formatted tasks
        output_file: Output filename
    """
    
    # Create batch structure
    batch_data = {
        "version": "1.0",
        "created_at": datetime.now().isoformat(),
        "project_config": {
            "title": "LULC Relationship Extraction",
            "description": "Comprehensive Land Use Land Cover relationship extraction and annotation",
            "label_config": get_labelstudio_config(),
            "expert_instruction": "Extract LULC transformations, spatial-temporal context, and causal relationships"
        },
        "tasks": results,
        "total_tasks": len(results),
        "statistics": {
            "relationships_found": sum(1 for r in results if r["data"]["meta"]["relationship_found"]),
            "total_entities": sum(r["data"]["meta"]["entity_count"] for r in results),
            "avg_entities_per_task": sum(r["data"]["meta"]["entity_count"] for r in results) / len(results) if results else 0
        }
    }
    
    # Save to file
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(batch_data, f, indent=2, ensure_ascii=False)
    
    print(f"✅ Label Studio batch saved to {output_file}")
    print(f"📊 Statistics:")
    print(f"  Total tasks: {batch_data['total_tasks']}")
    print(f"  Relationships found: {batch_data['statistics']['relationships_found']}")
    print(f"  Total entities: {batch_data['statistics']['total_entities']}")
    print(f"  Avg entities per task: {batch_data['statistics']['avg_entities_per_task']:.1f}")

def get_labelstudio_config() -> str:
    """Get Label Studio XML configuration for LULC relationship annotation."""
    
    config = """
    <View>
        <Header value="LULC Relationship Extraction"/>
        
        <!-- Text Display -->
        <Text name="text" value="$text"/>
        
        <!-- Entity Labeling -->
        <Labels name="label" toName="text">
            <Label value="LULC" background="green"/>
            <Label value="LOCATION" background="blue"/>
            <Label value="DATE" background="orange"/>
            <Label value="CHANGE" background="red"/>
            <Label value="PERCENT" background="purple"/>
            <Label value="MAGNITUDE" background="yellow"/>
            <Label value="PROCESS" background="pink"/>
            <Label value="DRIVER" background="brown"/>
        </Labels>
        
        <!-- Relationship Labeling -->
        <Relations name="relation" toName="label">
            <Relation value="TRANSFORMS_TO"/>
            <Relation value="CAUSES"/>
            <Relation value="LOCATED_IN"/>
            <Relation value="OCCURS_DURING"/>
            <Relation value="HAS_MAGNITUDE"/>
        </Relations>
        
        <!-- Document Classification -->
        <Choices name="document_type" toName="text" choice="single">
            <Choice value="LULC_CONVERSION"/>
            <Choice value="LULC_EXPANSION"/>
            <Choice value="LULC_DEGRADATION"/>
            <Choice value="LULC_RESTORATION"/>
            <Choice value="NO_LULC_RELATIONSHIP"/>
        </Choices>
        
        <!-- Spatial Scale -->
        <Choices name="spatial_scale" toName="text" choice="single">
            <Choice value="SCALE_LOCAL"/>
            <Choice value="SCALE_REGIONAL"/>
            <Choice value="SCALE_NATIONAL"/>
            <Choice value="SCALE_GLOBAL"/>
        </Choices>
        
        <!-- Structured Output -->
        <TextArea name="transformation_summary" toName="text" 
                 placeholder="Transformation: FROM -> TO (CHANGE_TYPE)"
                 maxSubmissions="1"/>
        
        <TextArea name="causal_chain" toName="text" 
                 placeholder="Causal chain: DRIVER -> TRANSFORMATION -> EFFECT"
                 maxSubmissions="1"/>
    </View>
    """
    
    return config.strip()

# Example usage function
def test_labelstudio_conversion():
    """Test the Label Studio conversion with sample data."""
    
    # Sample data
    sample_sentence = "Forest cover declined by 15% in the Amazon region between 2010 and 2020."
    sample_entities = [
        {"text": "Forest", "label": "LULC", "start": 0, "end": 6},
        {"text": "declined", "label": "CHANGE", "start": 13, "end": 21},
        {"text": "15%", "label": "PERCENT", "start": 25, "end": 28},
        {"text": "Amazon region", "label": "LOCATION", "start": 36, "end": 49},
        {"text": "2010", "label": "DATE", "start": 58, "end": 62},
        {"text": "2020", "label": "DATE", "start": 67, "end": 71}
    ]
    
    sample_relationships = {
        "relationship_found": True,
        "transformation": {
            "from_lulc": "forest",
            "to_lulc": "",
            "change_type": "decline",
            "magnitude": "15%"
        },
        "spatial_context": {
            "location": "Amazon region",
            "spatial_scale": "regional"
        },
        "temporal_context": {
            "start_date": "2010",
            "end_date": "2020"
        },
        "causal_mechanisms": {
            "primary_drivers": "deforestation"
        }
    }
    
    # Convert to Label Studio format
    ls_task = convert_to_labelstudio_format(
        sample_sentence, 
        sample_entities, 
        sample_relationships,
        task_id=1
    )
    
    # Save single task
    with open("sample_lulc_labelstudio.json", 'w', encoding='utf-8') as f:
        json.dump(ls_task, f, indent=2, ensure_ascii=False)
    
    print("✅ Sample Label Studio task created: sample_lulc_labelstudio.json")
    return ls_task

print("✅ Label Studio JSON format converter defined!")
print("📊 Functions available:")
print("- convert_to_labelstudio_format() - Main conversion function")
print("- save_labelstudio_batch() - Save multiple tasks")
print("- test_labelstudio_conversion() - Test with sample data")
print("- get_labelstudio_config() - Get XML configuration")
```
