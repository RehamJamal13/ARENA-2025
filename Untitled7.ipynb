```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json

def load_csv_data(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence"""
    
    df = pd.read_csv(csv_path)
    
    # Group data by sentence
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        # Skip empty rows (sentences with no relations)
        if pd.isna(row['source']) or row['source'] == '':
            continue
            
        # Extract entities
        source_entity = (row['source'], row['source_type'])
        target_entity = (row['target'], row['target_type'])
        
        sentence_data[sentence_id]['entities'].add(source_entity)
        sentence_data[sentence_id]['entities'].add(target_entity)
        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract relations (source, source_type, relation, target, target_type)
        relation = (row['source'], row['source_type'], row['relationship'], 
                   row['target'], row['target_type'])
        sentence_data[sentence_id]['relations'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_entity_errors(gt_entities: Set, pred_entities: Set) -> Dict:
    """Analyze entity-level errors"""
    
    errors = {
        'false_positives': list(pred_entities - gt_entities),
        'false_negatives': list(gt_entities - pred_entities),
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Group errors by entity type
    for entity, entity_type in errors['false_positives']:
        errors['by_type'][entity_type]['fp'].append((entity, entity_type))
        
    for entity, entity_type in errors['false_negatives']:
        errors['by_type'][entity_type]['fn'].append((entity, entity_type))
    
    return dict(errors)

def analyze_relation_errors(gt_relations: Set, pred_relations: Set) -> Dict:
    """Analyze relation-level errors"""
    
    errors = {
        'false_positives': list(pred_relations - gt_relations),
        'false_negatives': list(gt_relations - pred_relations),
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Group errors by relation type
    for relation in errors['false_positives']:
        relation_type = relation[2]
        errors['by_type'][relation_type]['fp'].append(relation)
        
    for relation in errors['false_negatives']:
        relation_type = relation[2]
        errors['by_type'][relation_type]['fn'].append(relation)
    
    return dict(errors)

def evaluate_csv_results(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files"""
    
    print("📂 Loading data...")
    gt_data = load_csv_data(ground_truth_csv)
    pred_data = load_csv_data(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results
    results = {
        'overall': {},
        'entity_metrics': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'entity_errors': [],
            'relation_errors': []
        }
    }
    
    # Collect all entities and relations
    all_gt_entities = set()
    all_pred_entities = set()
    all_gt_relations = set()
    all_pred_relations = set()
    
    # Collect entities and relations by type
    entity_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set()})
        pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set()})
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
            
        # Entity evaluation for this sentence
        gt_entities = gt_sentence['entities']
        pred_entities = pred_sentence['entities']
        
        entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(gt_entities, pred_entities)
        sentence_result.update({
            'entity_precision': entity_p,
            'entity_recall': entity_r, 
            'entity_f1': entity_f1,
            'entity_tp': entity_tp,
            'entity_fp': entity_fp,
            'entity_fn': entity_fn
        })
        
        # Relation evaluation for this sentence
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_entities.update(gt_entities)
        all_pred_entities.update(pred_entities)
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        
        # Group by entity type
        for entity, entity_type in gt_entities:
            entity_by_type[entity_type]['gt'].add((entity, entity_type))
        for entity, entity_type in pred_entities:
            entity_by_type[entity_type]['pred'].add((entity, entity_type))
            
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[2]  # relation type is at index 2
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[2]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall metrics...")
    
    # Overall entity metrics
    entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(all_gt_entities, all_pred_entities)
    results['overall']['entities'] = {
        'precision': entity_p, 'recall': entity_r, 'f1': entity_f1,
        'tp': entity_tp, 'fp': entity_fp, 'fn': entity_fn
    }
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall']['relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn
    }
    
    # Per-type entity metrics
    for entity_type, data in entity_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['entity_metrics'][entity_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis
    results['error_analysis']['entity_errors'] = analyze_entity_errors(all_gt_entities, all_pred_entities)
    results['error_analysis']['relation_errors'] = analyze_relation_errors(all_gt_relations, all_pred_relations)
    
    return results

def print_evaluation_report(results: Dict):
    """Print comprehensive evaluation report with robust error handling"""
    
    print("\n" + "="*60)
    print("🎯 LULC MODEL EVALUATION REPORT")
    print("="*60)
    
    # Overall performance
    overall = results['overall']
    print(f"\n📈 OVERALL PERFORMANCE:")
    print(f"   Entity    | P: {overall['entities']['precision']:.3f} | R: {overall['entities']['recall']:.3f} | F1: {overall['entities']['f1']:.3f}")
    print(f"   Relations | P: {overall['relations']['precision']:.3f} | R: {overall['relations']['recall']:.3f} | F1: {overall['relations']['f1']:.3f}")
    
    # Entity performance by type
    print(f"\n🏷️  ENTITY PERFORMANCE BY TYPE:")
    print(f"{'Type':<12} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 55)
    
    # Convert all keys to strings and sort
    entity_items = []
    for key, metrics in results['entity_metrics'].items():
        entity_items.append((str(key), metrics))
    
    # Sort by string representation of keys
    entity_items.sort(key=lambda x: x[0])
    
    for entity_type, metrics in entity_items:
        try:
            print(f"{entity_type:<12} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{entity_type:<12} | ERROR: {e}")
    
    # Relation performance by type
    print(f"\n🔗 RELATION PERFORMANCE BY TYPE:")
    print(f"{'Type':<15} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 60)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by string representation of keys
    relation_items.sort(key=lambda x: x[0])
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<15} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<15} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    print(f"   Entity false positives: {len(errors['entity_errors']['false_positives'])}")
    print(f"   Entity false negatives: {len(errors['entity_errors']['false_negatives'])}")
    print(f"   Relation false positives: {len(errors['relation_errors']['false_positives'])}")
    print(f"   Relation false negatives: {len(errors['relation_errors']['false_negatives'])}")
    
    # Show some example errors
    print(f"\n🔍 SAMPLE ENTITY ERRORS:")
    entity_errors = errors['entity_errors']
    if entity_errors['false_positives']:
        print(f"   False Positives (first 3): {entity_errors['false_positives'][:3]}")
    if entity_errors['false_negatives']:
        print(f"   False Negatives (first 3): {entity_errors['false_negatives'][:3]}")
        
    print(f"\n🔍 SAMPLE RELATION ERRORS:")
    relation_errors = errors['relation_errors']
    if relation_errors['false_positives']:
        print(f"   False Positives (first 3): {relation_errors['false_positives'][:3]}")
    if relation_errors['false_negatives']:
        print(f"   False Negatives (first 3): {relation_errors['false_negatives'][:3]}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "lulc_relations_EVALUATION.csv"  # Update this path
PREDICTIONS_CSV = "RE claudi 3shot  - RE CLAUDE.csv"  # Update this path

if __name__ == "__main__":
    try:
        # Run evaluation
        results = evaluate_csv_results(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print detailed report
        print_evaluation_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json

def load_ground_truth_csv(csv_path: str) -> Dict:
    """Load ground truth CSV data (relation extraction format)"""
    if csv_path.endswith('.xlsx'):
        df = pd.read_excel(csv_path)
    else:
        df = pd.read_csv(csv_path)
    
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        if pd.isna(row['source']) or row['source'] == '':
            continue
            
        source_entity = (row['source'], row['source_type'])
        target_entity = (row['target'], row['target_type'])
        
        sentence_data[sentence_id]['entities'].add(source_entity)
        sentence_data[sentence_id]['entities'].add(target_entity)
        sentence_data[sentence_id]['sentence'] = sentence_text
        
        relation = (row['source'], row['source_type'], row['relationship'], 
                   row['target'], row['target_type'])
        sentence_data[sentence_id]['relations'].add(relation)
    
    return dict(sentence_data)

def load_predictions_csv(csv_path: str) -> Dict:
    """Load predictions CSV data"""
    if csv_path.endswith('.xlsx'):
        df = pd.read_excel(csv_path)
    else:
        df = pd.read_csv(csv_path)
    
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        if 'source' in df.columns and 'target' in df.columns:  # Relation format
            if pd.isna(row['source']) or row['source'] == '':
                continue
            source_entity = (row['source'], row['source_type'])
            target_entity = (row['target'], row['target_type'])
            sentence_data[sentence_id]['entities'].add(source_entity)
            sentence_data[sentence_id]['entities'].add(target_entity)
            relation = (row['source'], row['source_type'], row['relationship'], 
                       row['target'], row['target_type'])
            sentence_data[sentence_id]['relations'].add(relation)
        else:  # Entity format
            if pd.isna(row['entity']) or row['entity'] == '':
                continue
            entity = (row['entity'], row['entity_type'])
            sentence_data[sentence_id]['entities'].add(entity)
        
        sentence_data[sentence_id]['sentence'] = sentence_text
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    tp = len(true_set & pred_set)
    fp = len(pred_set - true_set)
    fn = len(true_set - pred_set)
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_errors_by_sentence(gt_data: Dict, pred_data: Dict) -> Dict:
    """Analyze errors sentence by sentence"""
    
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    sentence_errors = {}
    
    for sentence_id in all_sentence_ids:
        gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set(), 'sentence': ''})
        pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set(), 'sentence': ''})
        
        # Entity errors for this sentence
        entity_fp = list(pred_sentence['entities'] - gt_sentence['entities'])
        entity_fn = list(gt_sentence['entities'] - pred_sentence['entities'])
        entity_tp = list(gt_sentence['entities'] & pred_sentence['entities'])
        
        # Relation errors for this sentence
        relation_fp = list(pred_sentence['relations'] - gt_sentence['relations'])
        relation_fn = list(gt_sentence['relations'] - pred_sentence['relations'])
        relation_tp = list(gt_sentence['relations'] & pred_sentence['relations'])
        
        # Get sentence text (prefer ground truth, fallback to prediction)
        sentence_text = gt_sentence.get('sentence', '') or pred_sentence.get('sentence', '')
        
        sentence_errors[sentence_id] = {
            'sentence_text': sentence_text,
            'entities': {
                'true_positives': entity_tp,
                'false_positives': entity_fp,
                'false_negatives': entity_fn,
                'ground_truth_total': len(gt_sentence['entities']),
                'predicted_total': len(pred_sentence['entities'])
            },
            'relations': {
                'true_positives': relation_tp,
                'false_positives': relation_fp,
                'false_negatives': relation_fn,
                'ground_truth_total': len(gt_sentence['relations']),
                'predicted_total': len(pred_sentence['relations'])
            }
        }
    
    return sentence_errors

def print_sentence_error_analysis(sentence_errors: Dict):
    """Print detailed error analysis by sentence"""
    
    print("\n" + "="*100)
    print("📝 SENTENCE-BY-SENTENCE ERROR ANALYSIS")
    print("="*100)
    
    # Sort sentences by error count (most problematic first)
    sorted_sentences = sorted(sentence_errors.items(), 
                            key=lambda x: len(x[1]['entities']['false_positives']) + 
                                        len(x[1]['entities']['false_negatives']) + 
                                        len(x[1]['relations']['false_positives']) + 
                                        len(x[1]['relations']['false_negatives']), 
                            reverse=True)
    
    for sentence_id, errors in sorted_sentences:
        entity_errors = errors['entities']
        relation_errors = errors['relations']
        
        # Skip sentences with no errors
        total_errors = (len(entity_errors['false_positives']) + len(entity_errors['false_negatives']) + 
                       len(relation_errors['false_positives']) + len(relation_errors['false_negatives']))
        
        if total_errors == 0:
            continue
            
        print(f"\n📄 SENTENCE ID: {sentence_id}")
        print(f"📝 TEXT: \"{errors['sentence_text'][:150]}{'...' if len(errors['sentence_text']) > 150 else ''}\"")
        print("-" * 90)
        
        # Entity analysis for this sentence
        print(f"🏷️  ENTITIES:")
        print(f"   ✅ Correct (TP): {len(entity_errors['true_positives'])}")
        print(f"   ❌ Wrong predictions (FP): {len(entity_errors['false_positives'])}")
        print(f"   ❓ Missed (FN): {len(entity_errors['false_negatives'])}")
        
        if entity_errors['false_positives']:
            print(f"   📍 False Positive Entities:")
            for i, (entity, entity_type) in enumerate(entity_errors['false_positives'], 1):
                print(f"      {i}. '{entity}' ({entity_type})")
        
        if entity_errors['false_negatives']:
            print(f"   📍 False Negative Entities:")
            for i, (entity, entity_type) in enumerate(entity_errors['false_negatives'], 1):
                print(f"      {i}. '{entity}' ({entity_type})")
        
        # Relation analysis for this sentence
        print(f"\n🔗 RELATIONS:")
        print(f"   ✅ Correct (TP): {len(relation_errors['true_positives'])}")
        print(f"   ❌ Wrong predictions (FP): {len(relation_errors['false_positives'])}")
        print(f"   ❓ Missed (FN): {len(relation_errors['false_negatives'])}")
        
        if relation_errors['false_positives']:
            print(f"   📍 False Positive Relations:")
            for i, relation in enumerate(relation_errors['false_positives'], 1):
                source, source_type, rel_type, target, target_type = relation
                print(f"      {i}. '{source}' ({source_type}) --[{rel_type}]--> '{target}' ({target_type})")
        
        if relation_errors['false_negatives']:
            print(f"   📍 False Negative Relations:")
            for i, relation in enumerate(relation_errors['false_negatives'], 1):
                source, source_type, rel_type, target, target_type = relation
                print(f"      {i}. '{source}' ({source_type}) --[{rel_type}]--> '{target}' ({target_type})")
        
        print("\n" + "="*90)

def create_sentence_summary_table(sentence_errors: Dict):
    """Create summary table showing error counts per sentence"""
    
    print("\n" + "="*100)
    print("📊 SENTENCE ERROR SUMMARY TABLE")
    print("="*100)
    
    print(f"{'Sentence ID':<12} | {'Entity FP':<9} | {'Entity FN':<9} | {'Relation FP':<11} | {'Relation FN':<11} | {'Total Errors':<12}")
    print("-" * 85)
    
    total_entity_fp = 0
    total_entity_fn = 0  
    total_relation_fp = 0
    total_relation_fn = 0
    
    # Sort by total error count
    sorted_sentences = sorted(sentence_errors.items(), 
                            key=lambda x: len(x[1]['entities']['false_positives']) + 
                                        len(x[1]['entities']['false_negatives']) + 
                                        len(x[1]['relations']['false_positives']) + 
                                        len(x[1]['relations']['false_negatives']), 
                            reverse=True)
    
    for sentence_id, errors in sorted_sentences:
        entity_fp = len(errors['entities']['false_positives'])
        entity_fn = len(errors['entities']['false_negatives'])
        relation_fp = len(errors['relations']['false_positives'])
        relation_fn = len(errors['relations']['false_negatives'])
        total_errors = entity_fp + entity_fn + relation_fp + relation_fn
        
        # Only show sentences with errors
        if total_errors > 0:
            print(f"{sentence_id:<12} | {entity_fp:<9} | {entity_fn:<9} | {relation_fp:<11} | {relation_fn:<11} | {total_errors:<12}")
            
            total_entity_fp += entity_fp
            total_entity_fn += entity_fn
            total_relation_fp += relation_fp
            total_relation_fn += relation_fn
    
    print("-" * 85)
    print(f"{'TOTAL':<12} | {total_entity_fp:<9} | {total_entity_fn:<9} | {total_relation_fp:<11} | {total_relation_fn:<11} | {total_entity_fp + total_entity_fn + total_relation_fp + total_relation_fn:<12}")

def show_perfect_sentences(sentence_errors: Dict):
    """Show sentences with no errors"""
    
    perfect_sentences = []
    for sentence_id, errors in sentence_errors.items():
        total_errors = (len(errors['entities']['false_positives']) + len(errors['entities']['false_negatives']) + 
                       len(errors['relations']['false_positives']) + len(errors['relations']['false_negatives']))
        if total_errors == 0:
            perfect_sentences.append((sentence_id, errors['sentence_text']))
    
    print(f"\n✨ PERFECT SENTENCES (No Errors): {len(perfect_sentences)}")
    print("-" * 60)
    
    for sentence_id, sentence_text in perfect_sentences[:10]:  # Show first 10
        print(f"   {sentence_id}: \"{sentence_text[:100]}{'...' if len(sentence_text) > 100 else ''}\"")
    
    if len(perfect_sentences) > 10:
        print(f"   ... and {len(perfect_sentences) - 10} more perfect sentences")

# Set your file paths
GROUND_TRUTH_CSV = "lulc_relations_EVALUATION.csv"
PREDICTIONS_CSV = "RE claudi 3shot  - RE CLAUDE.csv"

# Run the analysis
print("📂 Loading data...")
gt_data = load_ground_truth_csv(GROUND_TRUTH_CSV)
pred_data = load_predictions_csv(PREDICTIONS_CSV)

print(f"   Ground Truth: {len(gt_data)} sentences")
print(f"   Predictions: {len(pred_data)} sentences")

# Collect all entities and relations for overall metrics
all_gt_entities = set()
all_pred_entities = set()
all_gt_relations = set()
all_pred_relations = set()

all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())

for sentence_id in all_sentence_ids:
    gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set()})
    pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set()})
    
    all_gt_entities.update(gt_sentence['entities'])
    all_pred_entities.update(pred_sentence['entities'])
    all_gt_relations.update(gt_sentence['relations'])
    all_pred_relations.update(pred_sentence['relations'])

# Calculate overall metrics
entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(all_gt_entities, all_pred_entities)
relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)

# Print basic metrics
print(f"\n📈 OVERALL PERFORMANCE:")
print(f"   Entity    | P: {entity_p:.3f} | R: {entity_r:.3f} | F1: {entity_f1:.3f}")
print(f"   Relations | P: {relation_p:.3f} | R: {relation_r:.3f} | F1: {relation_f1:.3f}")

# Analyze errors by sentence
sentence_errors = analyze_errors_by_sentence(gt_data, pred_data)

# Print sentence-by-sentence analysis
create_sentence_summary_table(sentence_errors)
show_perfect_sentences(sentence_errors)
print_sentence_error_analysis(sentence_errors)
```

```python
import pandas as pd
import csv
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json

def load_csv_data(csv_path: str) -> Dict:
    """Load CSV data and organize by sentence"""
    
    df = pd.read_csv(csv_path)
    
    # Group data by sentence
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        # Skip empty rows (sentences with no relations)
        if pd.isna(row['source']) or row['source'] == '':
            continue
            
        # Extract entities
        source_entity = (row['source'], row['source_type'])
        target_entity = (row['target'], row['target_type'])
        
        sentence_data[sentence_id]['entities'].add(source_entity)
        sentence_data[sentence_id]['entities'].add(target_entity)
        sentence_data[sentence_id]['sentence'] = sentence_text
        
        # Extract relations (source, source_type, relation, target, target_type)
        relation = (row['source'], row['source_type'], row['relationship'], 
                   row['target'], row['target_type'])
        sentence_data[sentence_id]['relations'].add(relation)
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    
    tp = len(true_set & pred_set)  # True positives
    fp = len(pred_set - true_set)  # False positives
    fn = len(true_set - pred_set)  # False negatives
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_entity_errors(gt_entities: Set, pred_entities: Set) -> Dict:
    """Analyze entity-level errors"""
    
    errors = {
        'false_positives': list(pred_entities - gt_entities),
        'false_negatives': list(gt_entities - pred_entities),
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Group errors by entity type
    for entity, entity_type in errors['false_positives']:
        errors['by_type'][entity_type]['fp'].append((entity, entity_type))
        
    for entity, entity_type in errors['false_negatives']:
        errors['by_type'][entity_type]['fn'].append((entity, entity_type))
    
    return dict(errors)

def analyze_relation_errors(gt_relations: Set, pred_relations: Set) -> Dict:
    """Analyze relation-level errors"""
    
    errors = {
        'false_positives': list(pred_relations - gt_relations),
        'false_negatives': list(gt_relations - pred_relations),
        'by_type': defaultdict(lambda: {'fp': [], 'fn': []})
    }
    
    # Group errors by relation type
    for relation in errors['false_positives']:
        relation_type = relation[2]
        errors['by_type'][relation_type]['fp'].append(relation)
        
    for relation in errors['false_negatives']:
        relation_type = relation[2]
        errors['by_type'][relation_type]['fn'].append(relation)
    
    return dict(errors)

def evaluate_csv_results(ground_truth_csv: str, predictions_csv: str) -> Dict:
    """Complete evaluation using CSV files"""
    
    print("📂 Loading data...")
    gt_data = load_csv_data(ground_truth_csv)
    pred_data = load_csv_data(predictions_csv)
    
    print(f"   Ground Truth: {len(gt_data)} sentences")
    print(f"   Predictions: {len(pred_data)} sentences")
    
    # Initialize results
    results = {
        'overall': {},
        'entity_metrics': {},
        'relation_metrics': {},
        'per_sentence': [],
        'error_analysis': {
            'missing_sentences': [],
            'extra_sentences': [],
            'entity_errors': [],
            'relation_errors': []
        }
    }
    
    # Collect all entities and relations
    all_gt_entities = set()
    all_pred_entities = set()
    all_gt_relations = set()
    all_pred_relations = set()
    
    # Collect entities and relations by type
    entity_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    relation_by_type = defaultdict(lambda: {'gt': set(), 'pred': set()})
    
    # Get all sentence IDs from both datasets
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    
    print(f"\n🔍 Analyzing {len(all_sentence_ids)} sentences...")
    
    for sentence_id in all_sentence_ids:
        sentence_result = {'sentence_id': sentence_id}
        
        # Get data for this sentence
        gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set()})
        pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set()})
        
        # Check for missing/extra sentences
        if sentence_id not in gt_data:
            results['error_analysis']['extra_sentences'].append(sentence_id)
        if sentence_id not in pred_data:
            results['error_analysis']['missing_sentences'].append(sentence_id)
            
        # Entity evaluation for this sentence
        gt_entities = gt_sentence['entities']
        pred_entities = pred_sentence['entities']
        
        entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(gt_entities, pred_entities)
        sentence_result.update({
            'entity_precision': entity_p,
            'entity_recall': entity_r, 
            'entity_f1': entity_f1,
            'entity_tp': entity_tp,
            'entity_fp': entity_fp,
            'entity_fn': entity_fn
        })
        
        # Relation evaluation for this sentence
        gt_relations = gt_sentence['relations']
        pred_relations = pred_sentence['relations']
        
        relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(gt_relations, pred_relations)
        sentence_result.update({
            'relation_precision': relation_p,
            'relation_recall': relation_r,
            'relation_f1': relation_f1,
            'relation_tp': relation_tp,
            'relation_fp': relation_fp,
            'relation_fn': relation_fn
        })
        
        results['per_sentence'].append(sentence_result)
        
        # Accumulate for overall metrics
        all_gt_entities.update(gt_entities)
        all_pred_entities.update(pred_entities)
        all_gt_relations.update(gt_relations)
        all_pred_relations.update(pred_relations)
        
        # Group by entity type
        for entity, entity_type in gt_entities:
            entity_by_type[entity_type]['gt'].add((entity, entity_type))
        for entity, entity_type in pred_entities:
            entity_by_type[entity_type]['pred'].add((entity, entity_type))
            
        # Group by relation type
        for relation in gt_relations:
            relation_type = relation[2]  # relation type is at index 2
            relation_by_type[relation_type]['gt'].add(relation)
        for relation in pred_relations:
            relation_type = relation[2]
            relation_by_type[relation_type]['pred'].add(relation)
    
    # Calculate overall metrics
    print("\n📊 Calculating overall metrics...")
    
    # Overall entity metrics
    entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(all_gt_entities, all_pred_entities)
    results['overall']['entities'] = {
        'precision': entity_p, 'recall': entity_r, 'f1': entity_f1,
        'tp': entity_tp, 'fp': entity_fp, 'fn': entity_fn
    }
    
    # Overall relation metrics  
    relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)
    results['overall']['relations'] = {
        'precision': relation_p, 'recall': relation_r, 'f1': relation_f1,
        'tp': relation_tp, 'fp': relation_fp, 'fn': relation_fn
    }
    
    # Per-type entity metrics
    for entity_type, data in entity_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['entity_metrics'][entity_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Per-type relation metrics
    for relation_type, data in relation_by_type.items():
        p, r, f1, tp, fp, fn = calculate_metrics(data['gt'], data['pred'])
        results['relation_metrics'][relation_type] = {
            'precision': p, 'recall': r, 'f1': f1,
            'tp': tp, 'fp': fp, 'fn': fn,
            'support': len(data['gt'])
        }
    
    # Error analysis
    results['error_analysis']['entity_errors'] = analyze_entity_errors(all_gt_entities, all_pred_entities)
    results['error_analysis']['relation_errors'] = analyze_relation_errors(all_gt_relations, all_pred_relations)
    
    return results

def print_evaluation_report(results: Dict):
    """Print comprehensive evaluation report with robust error handling"""
    
    print("\n" + "="*60)
    print("🎯 LULC MODEL EVALUATION REPORT")
    print("="*60)
    
    # Overall performance
    overall = results['overall']
    print(f"\n📈 OVERALL PERFORMANCE:")
    print(f"   Entity    | P: {overall['entities']['precision']:.3f} | R: {overall['entities']['recall']:.3f} | F1: {overall['entities']['f1']:.3f}")
    print(f"   Relations | P: {overall['relations']['precision']:.3f} | R: {overall['relations']['recall']:.3f} | F1: {overall['relations']['f1']:.3f}")
    
    # Entity performance by type
    print(f"\n🏷️  ENTITY PERFORMANCE BY TYPE:")
    print(f"{'Type':<12} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 55)
    
    # Convert all keys to strings and sort
    entity_items = []
    for key, metrics in results['entity_metrics'].items():
        entity_items.append((str(key), metrics))
    
    # Sort by string representation of keys
    entity_items.sort(key=lambda x: x[0])
    
    for entity_type, metrics in entity_items:
        try:
            print(f"{entity_type:<12} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{entity_type:<12} | ERROR: {e}")
    
    # Relation performance by type
    print(f"\n🔗 RELATION PERFORMANCE BY TYPE:")
    print(f"{'Type':<15} | {'Precision':<9} | {'Recall':<9} | {'F1':<9} | {'Support':<8}")
    print("-" * 60)
    
    # Convert all keys to strings and sort
    relation_items = []
    for key, metrics in results['relation_metrics'].items():
        relation_items.append((str(key), metrics))
    
    # Sort by string representation of keys
    relation_items.sort(key=lambda x: x[0])
    
    for relation_type, metrics in relation_items:
        try:
            print(f"{relation_type:<15} | {metrics['precision']:<9.3f} | {metrics['recall']:<9.3f} | "
                  f"{metrics['f1']:<9.3f} | {metrics['support']:<8}")
        except (KeyError, TypeError) as e:
            print(f"{relation_type:<15} | ERROR: {e}")
    
    # Error summary
    errors = results['error_analysis']
    print(f"\n❌ ERROR SUMMARY:")
    print(f"   Missing sentences: {len(errors['missing_sentences'])}")
    print(f"   Extra sentences: {len(errors['extra_sentences'])}")
    print(f"   Entity false positives: {len(errors['entity_errors']['false_positives'])}")
    print(f"   Entity false negatives: {len(errors['entity_errors']['false_negatives'])}")
    print(f"   Relation false positives: {len(errors['relation_errors']['false_positives'])}")
    print(f"   Relation false negatives: {len(errors['relation_errors']['false_negatives'])}")
    
    # Show some example errors
    print(f"\n🔍 SAMPLE ENTITY ERRORS:")
    entity_errors = errors['entity_errors']
    if entity_errors['false_positives']:
        print(f"   False Positives (first 3): {entity_errors['false_positives'][:3]}")
    if entity_errors['false_negatives']:
        print(f"   False Negatives (first 3): {entity_errors['false_negatives'][:3]}")
        
    print(f"\n🔍 SAMPLE RELATION ERRORS:")
    relation_errors = errors['relation_errors']
    if relation_errors['false_positives']:
        print(f"   False Positives (first 3): {relation_errors['false_positives'][:3]}")
    if relation_errors['false_negatives']:
        print(f"   False Negatives (first 3): {relation_errors['false_negatives'][:3]}")

def convert_sets_to_lists(obj):
    """Convert sets to lists for JSON serialization"""
    if isinstance(obj, dict):
        return {key: convert_sets_to_lists(value) for key, value in obj.items()}
    elif isinstance(obj, list):
        return [convert_sets_to_lists(item) for item in obj]
    elif isinstance(obj, set):
        return list(obj)
    else:
        return obj

def save_results_to_json(results: Dict, output_path: str):
    """Save evaluation results to JSON file"""
    json_results = convert_sets_to_lists(results)
    with open(output_path, 'w') as f:
        json.dump(json_results, f, indent=2)
    print(f"\n💾 Results saved to: {output_path}")

# Set your file paths here
GROUND_TRUTH_CSV = "lulc_relations_EVALUATION.csv"  # Update this path
PREDICTIONS_CSV = "gemmeni ai pro 3 shot - ai_studio_code RE 3 SHOT.csv"  # Update this path

if __name__ == "__main__":
    try:
        # Run evaluation
        results = evaluate_csv_results(GROUND_TRUTH_CSV, PREDICTIONS_CSV)
        
        # Print detailed report
        print_evaluation_report(results)
        
        # Save results to JSON (optional)
        save_results_to_json(results, "evaluation_results.json")
        
    except FileNotFoundError as e:
        print(f"❌ Error: Could not find file - {e}")
        print("Please check your file paths and ensure the files exist.")
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
```

```python
import pandas as pd
from collections import defaultdict
from typing import Dict, List, Tuple, Set
import json

def load_ground_truth_csv(csv_path: str) -> Dict:
    """Load ground truth CSV data (relation extraction format)"""
    if csv_path.endswith('.xlsx'):
        df = pd.read_excel(csv_path)
    else:
        df = pd.read_csv(csv_path)
    
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        if pd.isna(row['source']) or row['source'] == '':
            continue
            
        source_entity = (row['source'], row['source_type'])
        target_entity = (row['target'], row['target_type'])
        
        sentence_data[sentence_id]['entities'].add(source_entity)
        sentence_data[sentence_id]['entities'].add(target_entity)
        sentence_data[sentence_id]['sentence'] = sentence_text
        
        relation = (row['source'], row['source_type'], row['relationship'], 
                   row['target'], row['target_type'])
        sentence_data[sentence_id]['relations'].add(relation)
    
    return dict(sentence_data)

def load_predictions_csv(csv_path: str) -> Dict:
    """Load predictions CSV data"""
    if csv_path.endswith('.xlsx'):
        df = pd.read_excel(csv_path)
    else:
        df = pd.read_csv(csv_path)
    
    sentence_data = defaultdict(lambda: {'entities': set(), 'relations': set()})
    
    for _, row in df.iterrows():
        sentence_id = row['sentence_id']
        sentence_text = row['sentence']
        
        if 'source' in df.columns and 'target' in df.columns:  # Relation format
            if pd.isna(row['source']) or row['source'] == '':
                continue
            source_entity = (row['source'], row['source_type'])
            target_entity = (row['target'], row['target_type'])
            sentence_data[sentence_id]['entities'].add(source_entity)
            sentence_data[sentence_id]['entities'].add(target_entity)
            relation = (row['source'], row['source_type'], row['relationship'], 
                       row['target'], row['target_type'])
            sentence_data[sentence_id]['relations'].add(relation)
        else:  # Entity format
            if pd.isna(row['entity']) or row['entity'] == '':
                continue
            entity = (row['entity'], row['entity_type'])
            sentence_data[sentence_id]['entities'].add(entity)
        
        sentence_data[sentence_id]['sentence'] = sentence_text
    
    return dict(sentence_data)

def calculate_metrics(true_set: Set, pred_set: Set) -> Tuple[float, float, float, int, int, int]:
    """Calculate precision, recall, F1 and counts"""
    tp = len(true_set & pred_set)
    fp = len(pred_set - true_set)
    fn = len(true_set - pred_set)
    
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    return precision, recall, f1, tp, fp, fn

def analyze_errors_by_sentence(gt_data: Dict, pred_data: Dict) -> Dict:
    """Analyze errors sentence by sentence"""
    
    all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())
    sentence_errors = {}
    
    for sentence_id in all_sentence_ids:
        gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set(), 'sentence': ''})
        pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set(), 'sentence': ''})
        
        # Entity errors for this sentence
        entity_fp = list(pred_sentence['entities'] - gt_sentence['entities'])
        entity_fn = list(gt_sentence['entities'] - pred_sentence['entities'])
        entity_tp = list(gt_sentence['entities'] & pred_sentence['entities'])
        
        # Relation errors for this sentence
        relation_fp = list(pred_sentence['relations'] - gt_sentence['relations'])
        relation_fn = list(gt_sentence['relations'] - pred_sentence['relations'])
        relation_tp = list(gt_sentence['relations'] & pred_sentence['relations'])
        
        # Get sentence text (prefer ground truth, fallback to prediction)
        sentence_text = gt_sentence.get('sentence', '') or pred_sentence.get('sentence', '')
        
        sentence_errors[sentence_id] = {
            'sentence_text': sentence_text,
            'entities': {
                'true_positives': entity_tp,
                'false_positives': entity_fp,
                'false_negatives': entity_fn,
                'ground_truth_total': len(gt_sentence['entities']),
                'predicted_total': len(pred_sentence['entities'])
            },
            'relations': {
                'true_positives': relation_tp,
                'false_positives': relation_fp,
                'false_negatives': relation_fn,
                'ground_truth_total': len(gt_sentence['relations']),
                'predicted_total': len(pred_sentence['relations'])
            }
        }
    
    return sentence_errors

def print_sentence_error_analysis(sentence_errors: Dict):
    """Print detailed error analysis by sentence"""
    
    print("\n" + "="*100)
    print("📝 SENTENCE-BY-SENTENCE ERROR ANALYSIS")
    print("="*100)
    
    # Sort sentences by error count (most problematic first)
    sorted_sentences = sorted(sentence_errors.items(), 
                            key=lambda x: len(x[1]['entities']['false_positives']) + 
                                        len(x[1]['entities']['false_negatives']) + 
                                        len(x[1]['relations']['false_positives']) + 
                                        len(x[1]['relations']['false_negatives']), 
                            reverse=True)
    
    for sentence_id, errors in sorted_sentences:
        entity_errors = errors['entities']
        relation_errors = errors['relations']
        
        # Skip sentences with no errors
        total_errors = (len(entity_errors['false_positives']) + len(entity_errors['false_negatives']) + 
                       len(relation_errors['false_positives']) + len(relation_errors['false_negatives']))
        
        if total_errors == 0:
            continue
            
        print(f"\n📄 SENTENCE ID: {sentence_id}")
        print(f"📝 TEXT: \"{errors['sentence_text'][:150]}{'...' if len(errors['sentence_text']) > 150 else ''}\"")
        print("-" * 90)
        
        # Entity analysis for this sentence
        print(f"🏷️  ENTITIES:")
        print(f"   ✅ Correct (TP): {len(entity_errors['true_positives'])}")
        print(f"   ❌ Wrong predictions (FP): {len(entity_errors['false_positives'])}")
        print(f"   ❓ Missed (FN): {len(entity_errors['false_negatives'])}")
        
        if entity_errors['false_positives']:
            print(f"   📍 False Positive Entities:")
            for i, (entity, entity_type) in enumerate(entity_errors['false_positives'], 1):
                print(f"      {i}. '{entity}' ({entity_type})")
        
        if entity_errors['false_negatives']:
            print(f"   📍 False Negative Entities:")
            for i, (entity, entity_type) in enumerate(entity_errors['false_negatives'], 1):
                print(f"      {i}. '{entity}' ({entity_type})")
        
        # Relation analysis for this sentence
        print(f"\n🔗 RELATIONS:")
        print(f"   ✅ Correct (TP): {len(relation_errors['true_positives'])}")
        print(f"   ❌ Wrong predictions (FP): {len(relation_errors['false_positives'])}")
        print(f"   ❓ Missed (FN): {len(relation_errors['false_negatives'])}")
        
        if relation_errors['false_positives']:
            print(f"   📍 False Positive Relations:")
            for i, relation in enumerate(relation_errors['false_positives'], 1):
                source, source_type, rel_type, target, target_type = relation
                print(f"      {i}. '{source}' ({source_type}) --[{rel_type}]--> '{target}' ({target_type})")
        
        if relation_errors['false_negatives']:
            print(f"   📍 False Negative Relations:")
            for i, relation in enumerate(relation_errors['false_negatives'], 1):
                source, source_type, rel_type, target, target_type = relation
                print(f"      {i}. '{source}' ({source_type}) --[{rel_type}]--> '{target}' ({target_type})")
        
        print("\n" + "="*90)

def create_sentence_summary_table(sentence_errors: Dict):
    """Create summary table showing error counts per sentence"""
    
    print("\n" + "="*100)
    print("📊 SENTENCE ERROR SUMMARY TABLE")
    print("="*100)
    
    print(f"{'Sentence ID':<12} | {'Entity FP':<9} | {'Entity FN':<9} | {'Relation FP':<11} | {'Relation FN':<11} | {'Total Errors':<12}")
    print("-" * 85)
    
    total_entity_fp = 0
    total_entity_fn = 0  
    total_relation_fp = 0
    total_relation_fn = 0
    
    # Sort by total error count
    sorted_sentences = sorted(sentence_errors.items(), 
                            key=lambda x: len(x[1]['entities']['false_positives']) + 
                                        len(x[1]['entities']['false_negatives']) + 
                                        len(x[1]['relations']['false_positives']) + 
                                        len(x[1]['relations']['false_negatives']), 
                            reverse=True)
    
    for sentence_id, errors in sorted_sentences:
        entity_fp = len(errors['entities']['false_positives'])
        entity_fn = len(errors['entities']['false_negatives'])
        relation_fp = len(errors['relations']['false_positives'])
        relation_fn = len(errors['relations']['false_negatives'])
        total_errors = entity_fp + entity_fn + relation_fp + relation_fn
        
        # Only show sentences with errors
        if total_errors > 0:
            print(f"{sentence_id:<12} | {entity_fp:<9} | {entity_fn:<9} | {relation_fp:<11} | {relation_fn:<11} | {total_errors:<12}")
            
            total_entity_fp += entity_fp
            total_entity_fn += entity_fn
            total_relation_fp += relation_fp
            total_relation_fn += relation_fn
    
    print("-" * 85)
    print(f"{'TOTAL':<12} | {total_entity_fp:<9} | {total_entity_fn:<9} | {total_relation_fp:<11} | {total_relation_fn:<11} | {total_entity_fp + total_entity_fn + total_relation_fp + total_relation_fn:<12}")

def show_perfect_sentences(sentence_errors: Dict):
    """Show sentences with no errors"""
    
    perfect_sentences = []
    for sentence_id, errors in sentence_errors.items():
        total_errors = (len(errors['entities']['false_positives']) + len(errors['entities']['false_negatives']) + 
                       len(errors['relations']['false_positives']) + len(errors['relations']['false_negatives']))
        if total_errors == 0:
            perfect_sentences.append((sentence_id, errors['sentence_text']))
    
    print(f"\n✨ PERFECT SENTENCES (No Errors): {len(perfect_sentences)}")
    print("-" * 60)
    
    for sentence_id, sentence_text in perfect_sentences[:10]:  # Show first 10
        print(f"   {sentence_id}: \"{sentence_text[:100]}{'...' if len(sentence_text) > 100 else ''}\"")
    
    if len(perfect_sentences) > 10:
        print(f"   ... and {len(perfect_sentences) - 10} more perfect sentences")

# Set your file paths
GROUND_TRUTH_CSV = "lulc_relations_EVALUATION.csv"
PREDICTIONS_CSV = "gemmeni ai pro 3 shot - ai_studio_code RE 3 SHOT.csv"

# Run the analysis
print("📂 Loading data...")
gt_data = load_ground_truth_csv(GROUND_TRUTH_CSV)
pred_data = load_predictions_csv(PREDICTIONS_CSV)

print(f"   Ground Truth: {len(gt_data)} sentences")
print(f"   Predictions: {len(pred_data)} sentences")

# Collect all entities and relations for overall metrics
all_gt_entities = set()
all_pred_entities = set()
all_gt_relations = set()
all_pred_relations = set()

all_sentence_ids = set(gt_data.keys()) | set(pred_data.keys())

for sentence_id in all_sentence_ids:
    gt_sentence = gt_data.get(sentence_id, {'entities': set(), 'relations': set()})
    pred_sentence = pred_data.get(sentence_id, {'entities': set(), 'relations': set()})
    
    all_gt_entities.update(gt_sentence['entities'])
    all_pred_entities.update(pred_sentence['entities'])
    all_gt_relations.update(gt_sentence['relations'])
    all_pred_relations.update(pred_sentence['relations'])

# Calculate overall metrics
entity_p, entity_r, entity_f1, entity_tp, entity_fp, entity_fn = calculate_metrics(all_gt_entities, all_pred_entities)
relation_p, relation_r, relation_f1, relation_tp, relation_fp, relation_fn = calculate_metrics(all_gt_relations, all_pred_relations)

# Print basic metrics
print(f"\n📈 OVERALL PERFORMANCE:")
print(f"   Entity    | P: {entity_p:.3f} | R: {entity_r:.3f} | F1: {entity_f1:.3f}")
print(f"   Relations | P: {relation_p:.3f} | R: {relation_r:.3f} | F1: {relation_f1:.3f}")

# Analyze errors by sentence
sentence_errors = analyze_errors_by_sentence(gt_data, pred_data)

# Print sentence-by-sentence analysis
create_sentence_summary_table(sentence_errors)
show_perfect_sentences(sentence_errors)
print_sentence_error_analysis(sentence_errors)
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score
from collections import defaultdict
import spacy
nlp = spacy.load("en_core_web_sm")
# File paths - update these with your actual file paths
ground_truth_path = 'GROUND TRUTH.csv'
model_output_path = 'extracted_entitiesCOT-3-LLAMA.csv'

# Load CSV files
print("Loading CSV files...")
ground_truth = pd.read_csv(ground_truth_path)
model_output = pd.read_csv(model_output_path)

# Display first few rows to verify structure
print("\nGround Truth Sample:")
print(ground_truth.head())
print("\nModel Output Sample:")
print(model_output.head())

def normalize_entity(entity):
    if pd.isna(entity):
        return ""
    entity = str(entity).lower().strip()
    doc = nlp(entity)
    lemmatized = " ".join([token.lemma_ for token in doc])
    return lemmatized

# Function to group entities by sentence with detailed printing
def group_entities_by_sentence(df, entity_column='entity', label_column='entity_type', 
                               sentence_id_column='sentence_id', show_process=True):
    """Group entities by sentence ID"""
    grouped = defaultdict(list)
    
    if show_process:
        print(f"\nGrouping entities from dataframe with {len(df)} rows...")
        print("-" * 80)
    
    for idx, row in df.iterrows():
        sentence_id = row[sentence_id_column]
        entity = row[entity_column]
        label = row[label_column]
        
        grouped[sentence_id].append({
            'entity': entity,
            'entity_normalized': normalize_entity(entity),  # Add normalized version
            'label': label
        })
        
        # Show first few groupings as examples
        if show_process and idx < 10:
            print(f"Row {idx}: Sentence ID '{sentence_id}' -> Entity: '{entity}', Label: '{label}'")
    
    if show_process:
        print("-" * 80)
        print(f"Total sentences with entities: {len(grouped)}")
        print(f"Total entities processed: {sum(len(entities) for entities in grouped.values())}")
        
        # Show example of grouped data
        print("\nExample of grouped data (first 3 sentences):")
        for i, (sent_id, entities) in enumerate(grouped.items()):
            if i >= 3:
                break
            print(f"\nSentence ID: {sent_id}")
            print(f"  Entities in this sentence: {len(entities)}")
            for ent in entities:
                print(f"    - Entity: '{ent['entity']}' (normalized: '{ent['entity_normalized']}'), Label: '{ent['label']}'")
    
    return grouped

# Function to evaluate entity type WITH NORMALIZATION
def evaluate_entity_type(ground_truth_grouped, model_output_grouped, target_entity_type):
    """Evaluate performance for a specific entity type"""
    true_positives = 0
    false_positives = 0
    false_negatives = 0
    
    all_sentence_ids = set(ground_truth_grouped.keys()) | set(model_output_grouped.keys())
    
    for sentence_id in all_sentence_ids:
        # Get entities for this sentence from both sources
        gt_entities = ground_truth_grouped.get(sentence_id, [])
        model_entities = model_output_grouped.get(sentence_id, [])
        
        # Filter for target entity type and use NORMALIZED entities
        gt_target_entities = set([e['entity_normalized'] for e in gt_entities 
                                 if e['label'] == target_entity_type])
        model_target_entities = set([e['entity_normalized'] for e in model_entities 
                                    if e['label'] == target_entity_type])
        
        # Calculate metrics
        true_positives += len(gt_target_entities & model_target_entities)
        false_positives += len(model_target_entities - gt_target_entities)
        false_negatives += len(gt_target_entities - model_target_entities)
    
    # Calculate precision, recall, and F1
    precision = true_positives / (true_positives + false_positives) if (true_positives + false_positives) > 0 else 0
    recall = true_positives / (true_positives + false_negatives) if (true_positives + false_negatives) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    return {
        'entity_type': target_entity_type,
        'true_positives': true_positives,
        'false_positives': false_positives,
        'false_negatives': false_negatives,
        'precision': precision,
        'recall': recall,
        'f1_score': f1
    }

# Function to get detailed comparison WITH NORMALIZATION
def get_detailed_comparison(ground_truth_grouped, model_output_grouped, target_entity_type):
    """Get detailed sentence-by-sentence comparison"""
    detailed_results = []
    
    all_sentence_ids = set(ground_truth_grouped.keys()) | set(model_output_grouped.keys())
    
    for sentence_id in sorted(all_sentence_ids):
        gt_entities = ground_truth_grouped.get(sentence_id, [])
        model_entities = model_output_grouped.get(sentence_id, [])
        
        # Create mappings from normalized to original for display
        gt_norm_to_orig = {e['entity_normalized']: e['entity'] 
                          for e in gt_entities if e['label'] == target_entity_type}
        model_norm_to_orig = {e['entity_normalized']: e['entity'] 
                             for e in model_entities if e['label'] == target_entity_type}
        
        gt_target = set(gt_norm_to_orig.keys())
        model_target = set(model_norm_to_orig.keys())
        
        # Find matches and mismatches
        correct = gt_target & model_target
        missed = gt_target - model_target
        false_pred = model_target - gt_target
        
        detailed_results.append({
            'sentence_id': sentence_id,
            'ground_truth_entities': [gt_norm_to_orig[e] for e in gt_target],
            'model_entities': [model_norm_to_orig[e] for e in model_target],
            'correct_predictions': [f"GT:'{gt_norm_to_orig[e]}' Model:'{model_norm_to_orig[e]}'" for e in correct],
            'missed_entities': [gt_norm_to_orig[e] for e in missed],
            'false_predictions': [model_norm_to_orig[e] for e in false_pred]
        })
    
    return detailed_results

# Group entities by sentence - SHOWING THE PROCESS
print("\n" + "="*80)
print("GROUPING GROUND TRUTH DATA")
print("="*80)
gt_grouped = group_entities_by_sentence(ground_truth, show_process=True)

print("\n" + "="*80)
print("GROUPING MODEL OUTPUT DATA")
print("="*80)
model_grouped = group_entities_by_sentence(model_output, show_process=True)

# Show statistics about entity types
print("\n" + "="*80)
print("ENTITY TYPE DISTRIBUTION")
print("="*80)

# Count entities by type in ground truth
gt_entity_counts = defaultdict(int)
for entities in gt_grouped.values():
    for ent in entities:
        gt_entity_counts[ent['label']] += 1

print("\nGround Truth Entity Counts:")
for label, count in gt_entity_counts.items():
    print(f"  {label}: {count}")

# Count entities by type in model output
model_entity_counts = defaultdict(int)
for entities in model_grouped.values():
    for ent in entities:
        model_entity_counts[ent['label']] += 1

print("\nModel Output Entity Counts:")
for label, count in model_entity_counts.items():
    print(f"  {label}: {count}")

# Entity types to evaluate
entity_types = ['LULC', 'PROCESS']

# Evaluate each entity type
results = {}
for entity_type in entity_types:
    print(f"\n{'='*50}")
    print(f"Evaluating {entity_type} (with normalization)")
    print('='*50)
    
    # Calculate metrics
    metrics = evaluate_entity_type(gt_grouped, model_grouped, entity_type)
    results[entity_type] = metrics
    
    # Print results
    print(f"True Positives: {metrics['true_positives']}")
    print(f"False Positives: {metrics['false_positives']}")
    print(f"False Negatives: {metrics['false_negatives']}")
    print(f"Precision: {metrics['precision']:.4f}")
    print(f"Recall: {metrics['recall']:.4f}")
    print(f"F1 Score: {metrics['f1_score']:.4f}")
    
    # Get detailed comparison
    detailed = get_detailed_comparison(gt_grouped, model_grouped, entity_type)
    
    # Save detailed results to CSV
    detailed_df = pd.DataFrame(detailed)
    detailed_df.to_csv(f'{entity_type}_detailed_comparison.csv', index=False)
    print(f"\nDetailed comparison saved to {entity_type}_detailed_comparison.csv")
    
    # Show some examples of errors
    print(f"\nExamples of sentences with errors for {entity_type}:")
    error_examples = [d for d in detailed if d['missed_entities'] or d['false_predictions']][:5]
    for example in error_examples:
        print(f"\nSentence ID: {example['sentence_id']}")
        if example['correct_predictions']:
            print(f"  Correct (case-different): {example['correct_predictions']}")
        if example['missed_entities']:
            print(f"  Missed: {example['missed_entities']}")
        if example['false_predictions']:
            print(f"  False predictions: {example['false_predictions']}")

# Save and display summary results
summary_df = pd.DataFrame(results).T
summary_df.to_csv('evaluation_summary.csv')
print("\n\nSummary results saved to evaluation_summary.csv")

# Overall summary
print("\n" + "="*50)
print("OVERALL SUMMARY (WITH NORMALIZATION)")
print("="*50)
print(summary_df[['precision', 'recall', 'f1_score']])
```
